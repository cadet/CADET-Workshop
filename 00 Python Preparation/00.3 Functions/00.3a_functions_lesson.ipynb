## Why functions matter

Functions help us:

1. **Avoid repetition** - Write once, use many times
2. **Organize code** - Break complex problems into smaller parts  
3. **Make debugging easier** - Test parts individually
4. **Avoid mistakes** - Reuse tested code instead of rewriting

## Defining functions - the basics

**Syntax**:
```python
def function_name(parameters):
    # do something
    return result  # optional
```

**Key Parts:**

- `def` - tells Python we're defining a function
- `function_name` - what we call our function
- `parameters` - input data (optional)
- `return` - output data (optional)

````python
# Basic function definitions - working with basic types

# 1. Simple function with no parameters
def greet():
    return "Hello, World!"

# 2. Function with parameters
def greet_person(name):
    return f"Hello, {name}!"

# 3. Function with multiple parameters
def greet_person_age(name, age):
    return f"Hello, {name}! You are {age} years old."

# 4. Function with default parameters
def greet_with_title(name, title="Friend"):
    return f"Hello, {title} {name}!"

# Using our functions
print("\nUsing our functions:")
print(f"  {greet()}")
print(f"  {greet_person('Anna')}")
print(f"  {greet_with_title('Clara')}")
print(f"  {greet_person_age('Bob', 25)}")
print(f"  {greet_with_title('David')}")
print(f"  {greet_with_title('Eve', title='Dr.')}")

#Q: What happens if we call greet_with_title with no "title" argument?
#print(f"  {greet_with_title('Eve', 'Dr.')}")
````

## Function scope

Function parameters and names assigned inside a function are **local** to that call. A function can read names from the surrounding notebook. Assigning to the same name inside the function normally creates a local variable instead.

The example shows how `global` explicitly rebinds a notebook-level name. For workshop calculations, passing inputs and returning results usually makes the dependencies clearer.

````python
# Global variables (accessible everywhere) - basic types
global_message = "I'm global!"      # str type
global_counter = 0                  # int type

def demonstrate_scope(parameter):
    # Local variables (only accessible inside this function)
    local_message = "I'm local!"     # str type
    local_number = 42                # int type
    
    # We can read global variables
    print(f"  Inside function - Global message: {global_message}")
    print(f"  Inside function - Global counter: {global_counter}")
    print(f"  Inside function - Local message: {local_message}")
    print(f"  Inside function - Local number: {local_number}")
    print(f"  Inside function - Parameter: {parameter}")


# Test scope with different types
print(f"Before function calls - Global message: {global_message}")
print(f"Before function calls - Global counter: {global_counter}")

#Q: What happens if we try to access local_number here?
#print(f"  Inside function - Local number: {local_number}")

def modify_global():
    global global_counter  # Tell Python we want to modify the global
    global_counter += 1
    return f"Counter is now: {global_counter}"


print(f"\nModifying global counter:")
print(f"  {modify_global()}")
print(f"  {modify_global()}")
print(f"  {modify_global()}")
````

## Return values

`return` passes a result back to the caller. Several returned values form a tuple, which can be unpacked into separate variables. A function without `return` returns `None`.

The examples show one result, several results and a conditional return. The final example returns a function as an additional pattern.

````python
# Return values and multiple returns
def simple_math(a, b):
    """Function returning a single value"""
    return a + b

def math_operations(a, b):
    """Function returning multiple values as a tuple"""
    addition = a + b
    subtraction = a - b
    multiplication = a * b

    return addition, subtraction, multiplication

def analyze_number(number):
    """Function returning different types based on analysis"""
    if not isinstance(number, (int, float)):
        return False, "Not a number"
    
    # Return: is_positive, description
    is_positive = number > 0
    description = f"{number} is {'positive' if is_positive else 'negative or zero'}"
    
    return is_positive, description

def safe_divide(a, b):
    """Function with different return types based on conditions"""
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        return None  # Return None for invalid input
    
    if b == 0:
        return None  # Return None for error
    
    return a / b  # Return number for success
````

````python
# Test different return patterns
result = simple_math(10, 5)
print(f"   simple_math(10, 5) = {result} (type: {type(result).__name__})")
````

````python
# Multiple return values
add, sub, mul = math_operations(10, 3)
print(f"   Addition: {add}, Subtraction: {sub}, Multiplication: {mul}")
````

````python
# Can also capture all as tuple
all_results = math_operations(8, 2)
print(f"   All results as tuple: {all_results} (type: {type(all_results).__name__})")
````

````python
is_pos, desc = analyze_number(42)
print(f"   Number 42: positive={is_pos}, description='{desc}'")

# Test with invalid input
result_invalid = analyze_number("not a number")
print(f"   Invalid input: {result_invalid}")

test_cases = [(10, 2), (10, 0), ("hello", 5)]
for a, b in test_cases:
    result = safe_divide(a, b)
    print(f"   safe_divide({a}, {b}) = {result} (type: {type(result).__name__})")

# Functions returning functions (advanced)
def create_multiplier(factor):
    """Function that returns another function"""
    def multiplier(x):
        return x * factor
    return multiplier

# Create specialized functions
double = create_multiplier(2)
triple = create_multiplier(3)

print(f"   double(5) = {double(5)}")
print(f"   triple(4) = {triple(4)}")
print(f"   double function type: {type(double).__name__}")
````

## Exercise 3: mathematical calculator

Create a calculator system that works with different numeric types:

Write a function that 

- Adds two numbers
- Subtracts two numbers
- Multiplies two numbers

````python
# Exercise 3: Mathematical Calculator

def add_numbers(a, b):
    """
    Adds two numbers.
    
    Parameters:
    a: First number
    b: Second number
    
    Returns:
    Sum of a and b
    """
    return a + b

def subtract_numbers(a, b):
    """
    Subtracts two numbers.
    
    Parameters:
    a: First number
    b: Second number
    
    Returns:
    Difference of a and b (a - b)
    """
    return a - b

def multiply_numbers(a, b):
    """
    Multiplies two numbers.
    
    Parameters:
    a: First number
    b: Second number
    
    Returns:
    Product of a and b
    """
    return a * b

# Test the calculator functions
print("Mathematical Calculator:")

# Test addition
result1 = add_numbers(10, 5)
print(f"10 + 5 = {result1}")

result2 = add_numbers(3.14, 2.86)
print(f"3.14 + 2.86 = {result2}")

# Test subtraction
result3 = subtract_numbers(20, 8)
print(f"20 - 8 = {result3}")

result4 = subtract_numbers(15.5, 3.2)
print(f"15.5 - 3.2 = {result4}")

# Test multiplication
result5 = multiply_numbers(6, 7)
print(f"6 * 7 = {result5}")

result6 = multiply_numbers(2.5, 4.0)
print(f"2.5 * 4.0 = {result6}")

# CADET-Process examples
print("\nCADET-Process calculations:")

# Column volume calculation: π * r² * h
radius = 0.5  # cm
height = 10.0  # cm
pi_approx = 3.14159

# Calculate area: π * r²
radius_squared = multiply_numbers(radius, radius)
area = multiply_numbers(pi_approx, radius_squared)
volume = multiply_numbers(area, height)
print(f"Column volume: π * {radius}² * {height} = {volume:.2f} cm³")

# Flow rate calculation: sum of two pumps
pump_a_flow = 1.2  # mL/min
pump_b_flow = 0.8  # mL/min
total_flow = add_numbers(pump_a_flow, pump_b_flow)
print(f"Total flow rate: {pump_a_flow} + {pump_b_flow} = {total_flow} mL/min")

# Concentration difference
initial_conc = 10.5  # mg/mL
final_conc = 7.3     # mg/mL
conc_decrease = subtract_numbers(initial_conc, final_conc)
print(f"Concentration decrease: {initial_conc} - {final_conc} = {conc_decrease} mg/mL")

# Buffer preparation: dilution factor
stock_conc = 1000    # mM
target_conc = 50     # mM
dilution_factor = multiply_numbers(stock_conc / target_conc, 1)
print(f"Dilution factor: {stock_conc}/{target_conc} = {dilution_factor}x")
````
