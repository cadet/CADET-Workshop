## Conditional statements - the foundation of decisions

**Conditionals** let programs choose different paths based on `bool` values.

**Syntax**:
```python
if condition:        # condition must evaluate to True/False
    # do this if True
elif other_condition:  # optional: check another condition
    # do this if other_condition is True
else:                 # optional: fallback
    # do this if all conditions are False
```

**Key Parts:**

- `if` - checks the first condition  
- `elif` - checks additional conditions (optional, can have multiple)
- `else` - handles all other cases (optional)
- **Indentation** - Python uses indentation to group code blocks

### Remember
When we introduced booleans, we saw how to create conditions using comparison operators:

- `==` - equal to
- `!=` - not equal to
- `<` - less than
- `>` - greater than
- `<=` - less than or equal
- `>=` - greater than or equal

### Use in CADET-Process
Conditionals are essential for chromatography logic:

- `if concentration > limit:` - Detect breakthrough
- `if pH < 7.0:` - Adjust buffer conditions
- `if flow_rate <= 0:` - Error handling for invalid parameters
- `if simulation_converged:` - Check if results are valid

````python
# Basic conditional statements - working with basic types

# 1. Simple if statement
age = 18
if age >= 18:
    print("You are an adult")

# 2. if-else statement  
temperature = 25
if temperature > 30:
    print("It's hot!")
else:
    print("It's pleasant!")

# 3. if-elif-else statement
score = 85
if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")  
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")

print("\nConditional examples:")
print(f"Age {age}: {'Adult' if age >= 18 else 'Minor'}")
print(f"Temperature {temperature}°C: {'Hot' if temperature > 30 else 'Pleasant'}")
print(f"Score {score}: Grade {'A' if score >= 90 else 'B' if score >= 80 else 'C' if score >= 70 else 'F'}")

# Working with different types in conditionals
name = ""
if name:  # Empty string is False
    print(f"Hello, {name}")
else:
    print("No name provided")
````

## Logical operators - combining conditions

**Logical operators** combine multiple conditions into one `bool` result:

**Operators:**

- `and` - True only if **both** conditions are True
- `or` - True if **at least one** condition is True  
- `not` - Reverses the boolean value (True becomes False, False becomes True)

**Truth Tables:**
```python
# AND truth table
True and True    →   True
True and False   →   False  
False and True   →   False
False and False  →   False

# OR truth table  
True or True     →   True
True or False    →   True
False or True    →   True
False or False   →   False

# NOT truth table
not True         →   False
not False        →   True
```

````python
# Logical operators in action

age = 25
has_license = True
is_citizen = True

can_rent = age >= 18 and has_license
can_vote = age >= 18 and is_citizen

print(f"Can rent car (age >= 18 AND has_license): {can_rent}")
print(f"Can vote (age >= 18 AND citizen): {can_vote}")
````

````python
# OR operator examples  
sunny = False
warm = True

good_park_day = sunny or warm
stay_inside = not sunny and not warm

print(f"Good day for park (sunny OR warm): {good_park_day}")
print(f"Stay inside (NOT sunny AND NOT warm): {stay_inside}")

# NOT operator examples
raining = True
print(f"Go outside (NOT raining): {not raining}")
print(f"Take umbrella (raining): {raining}")
````

````python
# Combining multiple conditions
temperature = 22
humidity = 60
air_quality = "good"

comfortable = (temperature >= 20 and temperature <= 25) and humidity < 70
print(f"Comfortable weather: {comfortable}")

good_conditions = (temperature > 15) and (humidity < 80) and (air_quality == "good")
print(f"Good outdoor conditions: {good_conditions}")
````

````python
# Type-based logical operations
text = "hello"
number = 42
empty_string = ""

string_text = bool(text) and type(text) == str
positive_integer = number > 0 and type(number) == int
empty_check = not bool(empty_string)  # Empty string is False-ish

print(f"Text exists and is string: {string_text}")
print(f"Number is positive integer: {positive_integer}")
print(f"Empty string check: {empty_check}")  # Empty string is False-ish
````

## Indentation and scope - the structure of control flow

**Python uses indentation to define code blocks** - this is fundamental to control flow!

### Why indentation matters
Other languages use `{}` brackets, but Use **four spaces** to indent each code block:
```python
# CORRECT - Proper indentation
if temperature > 30:
    print("It's hot!")        # 4 spaces indentation
    print("Stay hydrated!")   # Same level = same block
else:
    print("Nice weather!")    # 4 spaces indentation
    print("Great for walks!") # Same level = same block

# INCORRECT - Will cause IndentationError
if temperature > 30:
print("It's hot!")           # No indentation = ERROR!
```

### Indentation rules

1. **Consistent spacing**: Use **four spaces** per indentation level
2. **Same level = same block**: All lines at the same indentation level belong together
3. **Nested blocks**: Inner blocks need more indentation than outer blocks
4. **First line after `:` must be indented**: The colon `:` starts a new block


### Names in a code block

````python
# Demonstrating Scopes and Indentation in Control Structures

# Global scope variables
global_var = "I'm global - accessible everywhere"
user_age = 25

print(f"Global variable: {global_var}")
print(f"User age: {user_age}")

if user_age >= 18:
    # Local scope inside if block
    status = "adult"           # Created in local scope
    message = "You can vote!"  # Created in local scope
    print(f"    Status: {status}")
    print(f"    Message: {message}")
    print(f"    Global var accessible: {global_var}")  # Global accessible
else:
    # Different local scope inside else block
    #Q: What value has status here?
    #print(f"    Status: {status}")
    status = "minor"           # Different local variable
    message = "Wait a few years"
    print(f"    Status: {status}")
    print(f"    Message: {message}")
````

## For loops - repeating actions

**For loops** let us repeat actions for a specific number of times or over a collection of items.

**Basic Syntax:**
```python
for variable in sequence:
    # do something with variable
```

**Common Patterns:**

- `for i in range(n)` - repeat n times (i goes from 0 to n-1)
- `for i in range(start, stop)` - repeat from start to stop-1  
- `for char in string` - process each character in a string

**Key Concepts:**

- The **loop variable** takes on each value in the sequence  
- **Indentation** defines what code is inside the loop
- After the loop, the variable keeps the last value it had

````python
# For loops with basic types

print("\nCounting with range:")
for i in range(5):  # 0, 1, 2, 3, 4
    print(f"Count: {i}")

print(f"After loop, i = {i}")  # i keeps last value: 4
````

````python
# Range with start and stop
print("\nRange with start and stop:")
for num in range(3, 8):  # 3, 4, 5, 6, 7
    print(f"Number: {num}")
````

## Processing collections of characters - strings in loops

**Strings are sequences of characters** - perfect for demonstrating how loops work with collections!

### String processing patterns

When we loop through strings, we can:

- **Access each character individually** - `for char in string`
- **Track position and character** - `for i, char in enumerate(string)`
- **Process character properties** - check if vowel, digit, uppercase, etc.
- **Build new strings** - accumulate results based on conditions

### Character analysis with types
Each character in a string is also a `str` type, so we can:

- **Check character types**: `char.isdigit()`, `char.isalpha()`, `char.isupper()`
- **Transform characters**: `char.upper()`, `char.lower()`
- **Compare characters**: Use all comparison operators (`==`, `!=`, `<`, `>`)
- **Use in conditions**: Empty string `""` is `False`, any other string is `True`

### Common string + loop patterns
```python
# Pattern 1: Character analysis
for char in text:
    if char.lower() in "aeiou":
        # process vowels
        
# Pattern 2: Position tracking  
for i, char in enumerate(text):
    print(f"Position {i}: '{char}'")
    
# Pattern 3: Conditional processing
for char in text:
    if char.isdigit():
        # handle numbers
    elif char.isalpha():
        # handle letters
    else:
        # handle special characters
```

## While loops - conditional repetition

**While loops** repeat actions as long as a condition remains `True`.

**Syntax:**
```python
while condition:
    # do something
    # (usually modify variables to eventually make condition False)
```

**Key Differences from For Loops:**

- **For loops**: Know how many times to repeat ("repeat 5 times")
- **While loops**: Repeat until something changes ("repeat until user says stop")

**Important:** Always make sure the condition can eventually become `False`, or you'll have an infinite loop!

**Common Patterns:**

- Counters: Increment a variable until it reaches a target
- User input: Keep asking until valid input is received
- Searching: Keep looking until item is found or end is reached

````python
# While loops with basic types
print("\nCounting down with while:")
count = 5
while count > 0:
    print(f"Countdown: {count}")
    count -= 1  # Decrease count (IMPORTANT: modify the condition variable!)
print("Blast off!")
````

````python
#  Accumulating values until a target
total = 0
number = 1
target = 15

while total < target:
    print(f"Adding {number}, total will be: {total + number}")
    total += number
    number += 1

print(f"Final total: {total} (reached target {target})")
````

````python
# Using break to exit loops early
text = "Python Programming"
index = 0
vowels = "aeiouAEIOU"

while index < len(text):
    char = text[index]
    print(f"Character {index}: '{char}'")
    
    if char in vowels:
        print(f"  → Found vowel '{char}' at position {index}!")
        break  # Exit the loop early
    else:
        print(f"  → '{char}' is not a vowel, continuing...")
    
    index += 1
````

## Loop control - break and continue

**Loop control** statements let us fine-tune how loops behave:

**`break` statement:**

- **Exits** the loop immediately
- Useful when you've found what you're looking for
- Skips any remaining iterations

**`continue` statement:**

- **Skips** the rest of the current iteration
- Jumps to the next iteration immediately  
- Useful for skipping invalid or unwanted items

**When to use:**

- `break` → "I found what I need, stop looking"
- `continue` → "Skip this item, but keep processing others"

Both work with **for loops** and **while loops**.

````python
# Loop control: break and continue

# 1. Using BREAK to exit early - find first negative result

# Calculate squares and look for first negative difference
base_number = 10
target_square = 64  # 8²

for i in range(15):
    current_square = i * i
    difference = current_square - target_square
    
    print(f"Step {i}: {i}² = {current_square}, difference = {difference}")
    
    if difference >= 0:
        print(f"  → Found first non-negative difference: {difference} at step {i}")
        print("  → BREAKING out of loop")
        break
    else:
        print(f"  → Difference {difference} is negative, continuing...")

print("Loop finished")
````

````python
# 2. Using CONTINUE to skip certain calculations

multiplier_base = 7
result_sum = 0
calculation_count = 0

for factor in range(1, 11):  # 1 to 10
    print(f"\nFactor {factor}:")
    
    # Skip even factors
    if factor % 2 == 0:
        print(f"  → Factor {factor} is even, skipping")
        continue  # Skip to next iteration
    
    # This code only runs for odd factors
    result = multiplier_base * factor
    result_sum += result
    calculation_count += 1
    
    print(f"  → Processing: {multiplier_base} × {factor} = {result}")
    print(f"  → Running sum: {result_sum}")

print(f"Final sum of odd calculations: {result_sum}")
print(f"Total calculations performed: {calculation_count}")
````

````python
# Find perfect squares less than 100, but stop after finding 3 primes
prime_count = 0
square_sum = 0
max_primes = 3

for number in range(2, 100):
    print(f"\nTesting number: {number}")
    
    # Check if it's a perfect square (skip if it is)
    sqrt_num = int(number ** 0.5)
    if sqrt_num * sqrt_num == number:
        print(f"  → {number} is a perfect square ({sqrt_num}²), skipping")
        continue
    
    # Check if it's prime (simple primality test)
    is_prime = True
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break
    
    if is_prime:
        print(f"  → {number} is prime!")
        prime_count += 1
        square_sum += number * number
        
        # Stop after finding enough primes
        if prime_count >= max_primes:
            print(f"  → Found {max_primes} primes, stopping!")
            break
    else:
        print(f"  → {number} is not prime")

print(f"Sum of squares of first {prime_count} primes: {square_sum}")
````

## Exercise 1: classify temperatures

Write a function that classifies temperatures into categories:

- "Cold" for temperatures below 10°C
- "Warm" for temperatures between 10°C and 25°C
- "Hot" for temperatures above 25°C

````python
# Exercise 1: Classify Temperatures

def classify_temperature(temp):
    """Classify temperature into categories"""
    if temp < 10:
        return "Cold"
    elif temp <= 25:
        return "Warm"
    else:
        return "Hot"

# Test with various temperatures
temperatures = range(0,20,5)

print("Temperature Classification:")
for temp in temperatures:
    category = classify_temperature(temp)
    print(f"{temp}°C is {category}")

# CADET-Process relevant example
print(f"\n CADET examples:")
reactor_temp = 22  # °C - typical room temperature for chromatography
storage_temp = 4   # °C - typical protein storage temperature
heat_sterilization = 121  # °C - autoclave temperature

print(f"Reactor temperature {reactor_temp}°C: {classify_temperature(reactor_temp)}")
print(f"Storage temperature {storage_temp}°C: {classify_temperature(storage_temp)}")
print(f"Sterilization temperature {heat_sterilization}°C: {classify_temperature(heat_sterilization)}")

# Advanced: Check if temperature is suitable for protein work
def is_protein_safe(temp):
    """Check if temperature is safe for protein work"""
    if temp < 2:
        return "Too cold - proteins may freeze"
    elif temp <= 8:
        return "Good for storage"
    elif temp <= 25:
        return "Suitable for experiments"
    else:
        return "Too hot - proteins may denature"

print(f"\nProtein safety check:")
for temp in range(0,100,20):
    safety = is_protein_safe(temp)
    print(f"{temp}°C: {safety}")
````

## Exercise 2: classify numbers
Write a program that processes a list of numbers and:

- Prints "Even" for even numbers
- Prints "Odd" for odd numbers
- Skips negative numbers using `continue`

````python
# Exercise 2: Classify Numbers

def classify_numbers(numbers):
    """Process list of numbers and classify them"""
    print("Number Classification:")
    
    for num in numbers:
        # Skip negative numbers
        if num < 0:
            print(f"Skipping negative number: {num}")
            continue
            
        # Classify even/odd
        if num % 2 == 0:
            print(f"{num} is Even")
        else:
            print(f"{num} is Odd")

# Test with various numbers
test_numbers = range(-5,6)
classify_numbers(test_numbers)
````
