## Learning path

| Topic | What it prepares you to do |
|---|---|
| **00.2 Types and Objects** | Work with numbers, names and boolean settings |
| **00.3 Functions** | Pass arguments and use returned results |
| **00.4 Control Flow** | Read conditions and loops |
| **00.5 Containers** | Store component values in lists and select entries |
| **00.6 Methods** | Read calls such as `flow_sheet.add_unit(inlet)` |
| **00.7 Dictionaries** | Store named settings and use keyword arguments |
| **00.8 Classes** | Distinguish objects, attributes and methods |
| **00.9 Libraries and Environments** | Import libraries and work in the right environment |

## Getting started

Run cells from top to bottom with **Shift+Enter**. After editing code, rerun that cell and later cells that use its results. Try the short exercises; the class-design challenge is optional.
