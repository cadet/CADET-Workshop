````
# Our first class!
class Person:
    def __init__(self, name, age):
        """Constructor - called when creating new Person objects"""
        self.name = name      # Attribute (stores data)
        self.age = age        # Attribute (stores data)
    
    def greet(self):
        """Method - behavior specific to Person objects"""
        return f"Hello, I'm {self.name} and I'm {self.age} years old!"
    
    def have_birthday(self):
        """Method that changes the object's state"""
        self.age += 1
        return f"{self.name} is now {self.age} years old!"
````

````
# Create objects (instances) of the class
anna = Person("Anna", 28)
ben = Person("Ben", 35)

print(f"Anna: {type(anna)}")
print(f"Ben: {type(ben)}")
````

````
# Objects are different instances but same type
print(f"Same type? {type(anna) == type(ben)}")
print(f"Same object? {anna is ben}")
````

````
# Call methods (like with strings, list, dict etc... !)
print(f"\n{anna.greet()}")
print(f"{ben.greet()}")

# Change state
print(f"\n{anna.have_birthday()}")
print(f"{anna.greet()}")  # Age changed!
````

## Understanding class components

Every class has two main parts:

### Constructor (`__init__`)

- Called **automatically** when creating new objects
- Sets up the **initial state** (attributes)
- Initializes attributes such as a name and starting values

### Methods

- Functions that **belong to** the class
- Can access and modify the object's data through `self`
- Define what the object **can do**

````
# Exploring the components of a class
class Car:
    def __init__(self, brand, model, year):
        """Constructor - sets initial state"""
        print(f"Creating a new {brand} {model}...")
        self.brand = brand      # Attribute: brand name
        self.model = model      # Attribute: car model  
        self.year = year        # Attribute: year made
        self.mileage = 0        # Attribute: default value
        
    def drive(self, kilometers):
        """Method - defines behavior"""
        if kilometers > 0:
            self.mileage += kilometers
            return f"{self.brand} drove {kilometers}km. Total: {self.mileage}km"
        return "Distance must be positive"
    
    def info(self):
        """Method - returns information about the car"""
        return f"{self.year} {self.brand} {self.model} - {self.mileage}km driven"
````

````
# Create objects - constructor runs automatically
car1 = Car("BMW", "X3", 2022)
car2 = Car("Tesla", "Model S", 2023)


car1_info  = car1.info()
car2_info = car2.info()
print(f"\nCar 1 info: {car1_info}")
print(f"Car 2 info: {car2_info}")
````

````
# Use methods to change state
car1_drive = car1.drive(150)
car1_info_after = car1.info()

car2_drive = car2.drive(200)
car2_info_after = car2.info()

print(f"\n{car1_drive}")
print(f"{car2_drive}")

# Each object maintains its own state
print(f"\nAfter driving:")
print(f"Car 1: {car1_info_after}")
print(f"Car 2: {car2_info_after}")
````

## Refer to the current object with `self`

**self** allows methods to access and modify the object's attributes, ensuring that each object maintains its own state.
**`self`** is a **reference to the current instance** of the class. Think of it as the object saying "me" or "this specific instance".

When you call a method on an object:
```python
anna.greet()  # You call it like this
```

Python **automatically** translates this to:
```python
Person.greet(anna)  # Python actually calls it like this!
```

So `self` is just **the object that the method was called on**.

It is used in the configuration of methods in classes to access or modify the object's attributes.

````
# Understanding self with a BankAccount example
class BankAccount:
    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        self.balance = initial_balance
        print(f"Account created for {self.owner} with {self.balance}€")
    
    def deposit(self, amount):
        """self refers to the specific account object"""
        if amount > 0:
            self.balance += amount  # self.balance = THIS account's balance
            return f"{self.owner} deposited {amount}€. New balance: {self.balance}€"
        return "Amount must be positive"
    
    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            return f"{self.owner} withdrew {amount}€. New balance: {self.balance}€"
        return "Invalid amount or insufficient funds"
    
    def check_balance(self):
        return f"{self.owner}'s balance: {self.balance}€"
    
    def get_account_info():
        """Static method - doesn't depend on self"""
        print("\nThis is a general bank account. Use deposit() and withdraw() to manage funds.")
````

````
# Create separate accounts - each has its own data
anna_account = BankAccount("Anna", 1000)
ben_account = BankAccount("Ben", 500)
BankAccount.get_account_info()
````

````
#check initial balances
anna_balance = anna_account.check_balance()
ben_balance = ben_account.check_balance()

print(anna_balance)
print(ben_balance)
````

````
# Operations on Anna's account don't affect Ben's account

anna_deposit = anna_account.deposit(200)  # self = anna_account
ben_withdraw = ben_account.withdraw(100)  # self = ben_account
print(f"\n{anna_deposit}")  # self = anna_account
print(f"{ben_withdraw}")    # self = ben_account
````

````
print(f"\nFinal state:")
print(anna_account.check_balance())
print(ben_account.check_balance())
````

## Class attributes vs instance attributes

Classes can have two types of attributes:

- **Instance attributes**: Unique to each object (what we've seen so far)
- **Class attributes**: Shared by all objects of that class

````
# Class attributes vs Instance attributes
class Student:
    # Class attribute - shared by ALL Student objects
    school_name = "Python University"
    total_students = 0
    
    def __init__(self, name, age, major):
        # Instance attributes - unique to each Student object
        self.name = name        # Each student has their own name
        self.age = age          # Each student has their own age
        self.major = major      # Each student has their own major
        
        # Modify class attribute when new student is created
        Student.total_students += 1
    
    def introduce(self):
        return f"Hi, I'm {self.name}, studying {self.major} at {Student.school_name}"
    
    @classmethod
    def get_total_students(cls):
        """Class method - works with class attributes"""
        return f"Total students enrolled: {cls.total_students}"
````

````
# Access class attribute via class

# Create student objects
alice = Student("Alice", 20, "Computer Science")
bob = Student("Bob", 21, "Mathematics")  
charlie = Student("Charlie", 19, "Physics")


print("Individual introductions:")
print(alice.introduce())
print(bob.introduce())
print(charlie.introduce())
````

````
# Access class attribute via class
print(f"\nSchool info:")
print(f"School name: {Student.school_name}") #call class attribute not instance attribute
print(Student.get_total_students()) 
````

````
# All students share the same school_name
print(f"\nShared attribute check:")
print(f"Alice's school: {alice.school_name}")
print(f"Bob's school: {bob.school_name}")
print(f"Same school? {alice.school_name == bob.school_name}")
````

````
# Change class attribute affects all instances
Student.school_name = "Advanced Python Academy"
print(f"\nAfter school name change:")
print(alice.introduce())
print(bob.introduce())
````

## Class decorators

You've seen `@classmethod` in the code above. Let's understand what the `@` symbol means!

**Decorators** are special functions that **modify** or **enhance** other functions or methods. The `@` symbol is Python's way to apply decorators cleanly.

### What decorators do

Think of decorators like **labels** or **special instructions** that change how a function behaves:

- `@classmethod` → "This method works with the class, not instances"
- `@staticmethod` → "This method doesn't need class or instance data"
- `@property` → "Make this method act like an attribute"

### The three most important decorators for classes

#### `@classmethod` - methods that work with the class

- Can access **class attributes** (like `cls.total_students`)
- Called on the **class itself**, not instances
- First parameter is `cls` (the class), not `self`

#### `@staticmethod` - independent methods

- Don't need access to `self` or `cls`
- Like regular functions, but organized inside the class
- Useful for utility functions related to the class

#### `@property` - methods that act like attributes

- Call methods **without parentheses** `obj.method` instead of `obj.method()`
- Great for **computed attributes** and **data validation**
- Makes your class interface cleaner

Let's see all three in action!

````
# Complete Decorator Examples
class BankAccount:
    """Comprehensive example showing all three decorator types"""
    
    # Class attributes
    bank_name = "Python Bank"
    total_accounts = 0
    interest_rate = 0.02  # 2% annual interest
    
    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        self._balance = initial_balance  # Private attribute (by convention)
        self.account_number = BankAccount.total_accounts + 1
        BankAccount.total_accounts += 1
    
    # Regular instance method (uses self)
    def deposit(self, amount):
        """Regular method - needs instance data"""
        if amount > 0:
            self._balance += amount
            return f"Deposited {amount}€. New balance: {self._balance}€"
        return "Amount must be positive"
    
    # @property - Method acts like an attribute
    @property
    def balance(self):
        """Access balance like an attribute: account.balance (no parentheses!)"""
        return self._balance
    
    @property  
    def formatted_balance(self):
        """Computed attribute - calculated every time it's accessed"""
        return f"{self._balance:,.2f}€"
    
    @property
    def account_info(self):
        """Complex computed attribute combining multiple data"""
        return f"Account #{self.account_number}: {self.owner} - {self.formatted_balance}"
    
    # @classmethod - Works with class data, not instance data
    @classmethod
    def get_bank_info(cls):
        """Class method - access with BankAccount.get_bank_info()"""
        return f"{cls.bank_name} - {cls.total_accounts} accounts, {cls.interest_rate*100}% interest rate"
    
    @classmethod
    def create_savings_account(cls, owner, initial_deposit=100):
        """Alternative constructor using @classmethod"""
        account = cls(owner, initial_deposit)
        account.account_type = "Savings"
        return account
    
    @classmethod
    def set_interest_rate(cls, new_rate):
        """Modify class-wide settings"""
        old_rate = cls.interest_rate
        cls.interest_rate = new_rate
        return f"Interest rate changed from {old_rate*100}% to {new_rate*100}%"
    
    # @staticmethod - Independent utility functions
    @staticmethod
    def validate_amount(amount):
        """Static method - doesn't need self or cls"""
        return isinstance(amount, (int, float)) and amount > 0
    
    @staticmethod
    def calculate_compound_interest(principal, rate, time):
        """Utility function related to banking"""
        return principal * (1 + rate) ** time
    
    @staticmethod
    def format_currency(amount):
        """Helper function for formatting"""
        return f"{amount:,.2f}€"
    
    def __str__(self):
        return f"BankAccount({self.owner}: {self.formatted_balance})"
````

````
# Test all decorator types
# Create accounts
account1 = BankAccount("Anna", 1000)
account2 = BankAccount("Ben", 500)

account1_deposit = account1.deposit(200)
account2_deposit = account2.deposit(100)

print(account1_deposit)
print(account2_deposit)
````

````
# 1. @property - METHODS THAT ACT LIKE ATTRIBUTES

account1_balance = account1.balance
account2_formatted = account2.formatted_balance
account1_info = account1.account_info

print(f"Anna's balance: {account1_balance}")            # No parentheses!
print(f"Ben's formatted balance: {account2_formatted}") # No parentheses!
print(f"Anna's info: {account1_info}")                  # No parentheses!
````

````
# 2.@classmethod - METHODS THAT WORK WITH THE CLASS ITSELF:
bankaccount_get_info = BankAccount.get_bank_info()      #Called on class

print(bankaccount_get_info)                             # Called on class
print(BankAccount.set_interest_rate(0.025))             # Modify class data

# Alternative constructor with @classmethod
savings = BankAccount.create_savings_account("Clara", 250)
print(f"New savings account: {savings}")
````

````
# 4. @staticmethod - INDEPENDENT UTILITY METHODS:
print(f"Is 100 valid amount? {BankAccount.validate_amount(100)}")
print(f"Is -50 valid amount? {BankAccount.validate_amount(-50)}")
````

````
# Calculate compound interest
future_value = BankAccount.calculate_compound_interest(1000, 0.02, 5)
print(f"1000€ at 2% for 5 years: {BankAccount.format_currency(future_value)}")
````

## Special methods

Python has **special methods** that connect class behavior with built-in functions:
**Magic methods** (also called **dunder methods** - "double underscore") are special methods that Python calls automatically when you use built-in functions and operators on your objects.

### Why "magic"?
They're called "magic" because **you don't call them directly** - Python calls them for you! When you write `print(obj)`, Python automatically calls `obj.__str__()`.

### The pattern: double underscores
All magic methods follow the pattern `__methodname__()`:

- `__init__()` - Called when creating objects
- `__str__()` - Called by `print()` and `str()`
- `__len__()` - Called by `len()`
- `__add__()` - Called by `+` operator

### How they work

**Without magic methods:**
```python
class SimpleClass:
    def __init__(self, value):
        self.value = value

obj = SimpleClass(42)
print(obj)  # Output: <__main__.SimpleClass object at 0x...>
```

**With magic methods:**
```python
class MagicClass:
    def __init__(self, value):
        self.value = value
    
    def __str__(self):
        return f"MagicClass with value: {self.value}"

obj = MagicClass(42)
print(obj)  # Output: MagicClass with value: 42
```

### Key insight
Magic methods make your custom classes behave like **built-in Python types**:

- Just like `len("hello")` works, you can make `len(your_object)` work
- Just like `str1 + str2` works, you can make `obj1 + obj2` work
- Just like `print(42)` works, you can make `print(your_object)` work beautifully

This is what makes Python classes so powerful and intuitive to use!

````
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def __str__(self):
        """Called by print() and str()"""
        return f"Account of {self.owner}: {self.balance:.2f}€"
    
    def __repr__(self):
        """Called by repr() and in lists"""
        return f"BankAccount('{self.owner}', {self.balance})"
    
    def __len__(self):
        """Called by len() - here: number of digits in balance"""
        return len(str(int(abs(self.balance))))
    
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            return f"{amount:.2f}€ deposited"
        return "Amount must be positive"
    
    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            return f"{amount:.2f}€ withdrawn"
        return "Invalid amount or insufficient funds"
````

````
# Magic Methods in action
account = BankAccount("Anna Miller", 1250.75)

print(f"print(): {account}")           # Calls __str__
print(f"repr(): {repr(account)}")       # Calls __repr__
print(f"len(): {len(account)} digits") # Calls __len__
````

````
# Regular methods
print(f"\n{account.deposit(500)}")
print(f"{account}")
print(f"{account.withdraw(200)}")
print(f"{account}")
````

## Additional example: inheritance

One of the most powerful features of classes is **inheritance** - the ability to create new classes based on existing ones. It's like creating specialized versions of something you already have!

### The concept: parent and child classes

Think of it like this:

- **Parent class** (base class): Contains common features
- **Child class** (derived class): Inherits everything from parent + adds new features

**Real-world analogy**: 

- `Vehicle` (parent) → has wheels, engine, can move
- `Car` (child) → inherits all Vehicle features + has 4 wheels, doors
- `Motorcycle` (child) → inherits all Vehicle features + has 2 wheels, no doors

### Key benefits of inheritance

**1. Code Reuse** 

- Write common functionality once in the parent class
- All child classes automatically get these features

**2. Polymorphism**

- Different classes can implement the same method differently
- You can treat all objects the same way (like calling `.work()` on any employee)

**3. Extensibility**

- Easy to add new types without changing existing code
- Just create a new child class

**4. Maintenance**

- Change shared behavior in one place (parent class)
- All child classes automatically get the updates

### When to use inheritance

**Good for inheritance:**

- Clear "is-a" relationship (Dog IS-A Animal, Developer IS-A Employee)
- Shared behavior between related classes
- Need to extend existing functionality

**Consider composition instead:**

- "has-a" relationship (Car HAS-A Engine, not Car IS-A Engine)
- Unrelated classes that just share some methods
- Complex multiple inheritance scenarios

````
# Basic inheritance example - Animal family

class Animal:  # Parent class
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def speak(self):
        return "Some generic animal sound"
    
    def info(self):
        return f"{self.name} is {self.age} years old"
    
    def sleep(self):
        return f"{self.name} is sleeping... "

# Child classes inherit from Animal
class Dog(Animal):  # Dog inherits from Animal
    def speak(self):  # Override parent method
        return f"{self.name} says: Woof! "
    
    def fetch(self):  # New method only for dogs
        return f"{self.name} is fetching the ball!"

class Cat(Animal):  # Cat inherits from Animal
    def speak(self):  # Override parent method
        return f"{self.name} says: Meow!"
    
    def climb(self):  # New method only for cats
        return f"{self.name} is climbing a tree!"
````

````
# Create instances
dog = Dog("Buddy", 3)
cat = Cat("Whiskers", 2)

# Demonstrate inheritance and method overriding
dog_info  = dog.info()
dog_speak = dog.speak()
dog_fetch = dog.fetch()
dog_sleep = dog.sleep()

print("DOG:")
print(dog_info)      # Inherited from Animal
print(dog_speak)     # Overridden in Dog
print(dog_fetch)     # Specific to Dog
print(dog_sleep)     # Inherited from Animal

print("\n CAT:")
print(cat.info())      # Inherited from Animal  
print(cat.speak())     # Overridden in Cat
print(cat.climb())     # Specific to Cat
print(cat.sleep())     # Inherited from Animal
````

````
# Both are still Animals!
is_animal_dog = isinstance(dog, Animal)
is_animal_cat = isinstance(cat, Animal)

print(f"\nIs dog an Animal? {is_animal_dog}")
print(f"Is cat an Animal? {is_animal_cat}")
print(f"Is dog a Dog? {isinstance(dog, Dog)}")
print(f"Is dog a Cat? {isinstance(dog, Cat)}")
````

### Advanced inheritance - calling parent methods

Sometimes you want to **extend** parent behavior instead of completely replacing it. Use `super()` to call parent methods:

````
# Advanced inheritance with super()

class Vehicle:  # Base class
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
        self.mileage = 0
    
    def start(self):
        return f"{self.brand} {self.model} is starting... 🔧"
    
    def drive(self, distance):
        self.mileage += distance
        return f"Drove {distance}km. Total mileage: {self.mileage}km"
    
    def info(self):
        return f"{self.year} {self.brand} {self.model} - {self.mileage}km"

class ElectricCar(Vehicle):  # Electric car inherits from Vehicle
    def __init__(self, brand, model, year, battery_capacity):
        super().__init__(brand, model, year)  # Call parent constructor
        self.battery_capacity = battery_capacity
        self.battery_level = 100  # Start fully charged
    
    def start(self):
        parent_message = super().start()  # Call parent start method
        return f"{parent_message} (Electric motor - silent!)"
    
    def charge(self, amount):
        old_level = self.battery_level
        self.battery_level = min(100, self.battery_level + amount)
        return f"Charged from {old_level}% to {self.battery_level}%"
    
    def info(self):
        parent_info = super().info()  # Call parent info
        return f"{parent_info} | Battery: {self.battery_level}% ({self.battery_capacity}kWh)"

class SportsCar(Vehicle):  # Sports car inherits from Vehicle
    def __init__(self, brand, model, year, top_speed):
        super().__init__(brand, model, year)
        self.top_speed = top_speed
    
    def start(self):
        parent_message = super().start()
        return f"{parent_message} (V8 engine roaring!)"
    
    def turbo_boost(self):
        return f"{self.brand} {self.model} activates turbo boost! (Top speed: {self.top_speed}km/h)"

# Test inheritance with super()
tesla = ElectricCar("Tesla", "Model 3", 2023, 75)
ferrari = SportsCar("Ferrari", "F8", 2022, 340)
````

````
print("ELECTRIC CAR:")
print(tesla.start())        # Extended start method
print(tesla.drive(150))     # Inherited method
print(tesla.charge(20))     # Unique method
print(tesla.info())         # Extended info method

print("\n SPORTS CAR:")
ferrari_start = ferrari.start()
ferrari_info = ferrari.info()
ferrari_drive = ferrari.drive(200)
ferrari_turbo = ferrari.turbo_boost()

print(ferrari_start)      # Extended start method  
print(ferrari_drive)   # Inherited method
print(ferrari_turbo) # Unique method
print(ferrari_info)       # Inherited method

# Show the inheritance hierarchy
print(f"\nTesla is a Vehicle: {isinstance(tesla, Vehicle)}")
print(f"Tesla is an ElectricCar: {isinstance(tesla, ElectricCar)}")
print(f"Ferrari is a Vehicle: {isinstance(ferrari, Vehicle)}")
print(f"Ferrari is a SportsCar: {isinstance(ferrari, SportsCar)}")
````

## Excursion: From Dictionaries to Classes - A Philosophy Shift

You've learned that dictionaries organize **data**:
```python
person_data = {"name": "Anna", "age": 28, "city": "Berlin"}
```

But what about **behavior**? What if Anna wants to:

- Introduce herself
- Have a birthday
- Move to a new city

**Classes solve this** by combining **data AND behavior** in one elegant package!

### The evolution of programming

1. **Variables**: Store single values (`name = "Anna"`)
2. **Lists**: Store multiple values (`names = ["Anna", "Ben"]`)
3. **Dictionaries**: Store named data (`person = {"name": "Anna", "age": 28}`)
4. **Classes**: Store data AND behavior (`person.greet()`, `person.have_birthday()`)

## Classes and dictionaries

Now let's understand when to use dictionaries and when to use classes. Both can store data, but they serve different purposes!

#### Dictionary approach vs. class approach

**Dictionary with a function:**
```python
person = {
    "name": "Anna", 
    "age": 28,
    "greet": lambda self: f"Hello, I'm {self['name']} and I'm {self['age']} years old!"
}
# Call the function from the dictionary
person["greet"](person)  # Must pass the dictionary as parameter
```

**Class with a method:**
```python
person = Person("Anna", 28)
person.greet()  # Method is part of the object!
```

#### Key advantages of classes

#### Data and behavior belong together

- Just like strings have `.upper()` method built-in!
- No need to pass the object to its own functions
- Methods automatically have access to the object's data

#### Structure enforcement
Classes have a **constructor** (`__init__`) that ensures consistent object creation:

```python
# Dictionary - no structure enforcement
person1 = {"name": "Anna"}      # Missing age? Valid but incomplete!
person2 = {"age": 28}           # Missing name? Also valid but problematic!
person3 = {"nam": "Bob"}        # Typo in key? No error, just wrong data!

# Class - structure is guaranteed
person1 = Person("Anna", 28)    # All required data must be provided
# person2 = Person("Anna")      # Error! Age is required
# person3 = Person()            # Error! Both name and age required
```

#### Type identity
```python
# Dictionaries are just dictionaries
person_dict = {"name": "Anna", "age": 28}
print(type(person_dict).__name__)  # dict

# Classes create custom types
person_obj = Person("Anna", 28)
print(type(person_obj).__name__)   # Person
```

#### Controlled flexibility vs. maximum freedom

**Dictionary - Maximum flexibility, less safety:**
```python
person_dict["salary"] = 50000           # Can add any key at runtime
person_dict["age"] = "twenty-eight"     # Can change type of values
del person_dict["name"]                 # Can remove essential data
person_dict["height"] = "tall"          # No validation of data
```

**Class - Controlled flexibility, more safety:**
```python
person_obj.salary = 50000               # Ordinary classes can accept new attributes
# person_obj.age = "twenty-eight"       # Type checking possible
# del person_obj.name                   # Can be prevented with properties
# person_obj.invalid_attr = "test"      # Can restrict new attributes
```

### When to use what?

| Use **Dictionaries** for: | Use **Classes** for: |
|---------------------------|---------------------|
|  Configuration data |  Business logic |
|  API responses |  Reusable components |
|  JSON-like structures |  Systems with behavior |
|  Quick prototypes |  Long-term projects |
|  Simple data storage |  Data with validation |

### The big picture

Think of it this way:

- **Dictionaries** = A flexible container (like a backpack)
- **Classes** = A specialized tool (like a smartphone with specific functions)

## Exercise 1: create your first class

Create a `Book` class for a library:

- Attributes: `title`, `author`, `year`, `checked_out` (default `False`)
- Methods: 
  - `check_out()`: Marks the book as checked out
  - `return_book()`: Marks the book as returned
  - `__str__()`: Returns a string representation of the book

````
# Exercise 1: Create Book Class

class Book:
    """Represents a book with title, author, ISBN and availability status"""
    
    def __init__(self, title, author, isbn):
        """Initialize a new book"""
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = True  # New books are available by default
    
    def check_out(self):
        """Mark the book as checked out"""
        if self.is_available:
            self.is_available = False
            return f"'{self.title}' has been checked out"
        else:
            return f"'{self.title}' is already checked out"
    
    def return_book(self):
        """Mark the book as returned"""
        if not self.is_available:
            self.is_available = True
            return f"'{self.title}' has been returned"
        else:
            return f"'{self.title}' was not checked out"
    
    def __str__(self):
        """String representation of the book"""
        status = "Available" if self.is_available else "Checked out"
        return f"'{self.title}' by {self.author} (ISBN: {self.isbn}) - {status}"
````

````
# Test the Book class

# Create some books
book1 = Book("Python Programming", "John Smith", "978-3-16-148410-0")
book2 = Book("Data Science Handbook", "Jane Doe", "978-1-86197-876-9")
book3 = Book("CADET Process Guide", "Dr. Mueller", "978-0-123456-78-9")

# Display books
print("Initial books:")
print(book1)
print(book2)
print(book3)
print()
````

````
# Test check out functionality
print("Testing check out:")
print(book1.check_out())
print(book1.check_out())  # Try to check out again
print(book1)
print()
````
