## How to read these notebooks

Each one follows the same four steps, which is also the order worth using on
your own failing simulation:

1. **Read the error message for what it does say.** The time of failure, the
   step size, and any warning printed before it are usually more informative
   than the error code itself.
2. **Change one thing at a time.** Discretization, tolerances, initialization
   mode, a single model parameter. The pattern of what helps and what does not
   narrows the cause down faster than any single clever guess.
3. **Evaluate your model equations by hand** at a state the simulation
   actually reaches. Solver settings cannot repair a term that is already
   meaningless in double precision.
4. **Ask whether the answer changes.** If a parameter can be moved over orders
   of magnitude without the result moving, it is not doing what you think it
   is doing.

All cases were reproduced with **CADET-Core v6.0.0a5**. Where the behaviour
differs from what the forum thread reported at the time, the notebook says so.
