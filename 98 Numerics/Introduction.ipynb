## Selecting the model

- Chromatographic processes can be descried by systems of **Partial differential (algebraic) equations (PD(A)E)**.
- Different models represent different physical phenomena and levels of complexity:
    - **Dimensionality:** 1D, 2D, 3D 
    - **Flow Geometry:** Axial, radial
    - **Transport:** Convection, dispersion, film and pore diffusion 
    - **Binding:** Linear and nonlinear adsorption
    - **Reactions:** Mass-action law, Michaelis Menten
- The governing equations depend on the modeling assumptions and the considered physical mechanisms.
- **CADET-Core** covers a vast model family
    - Governing equations and assumptions: [<u>CADET-Equations</u>](https://cadet-equations-74ko8eryoxmsbqspggxrj2.streamlit.app/)
    - Further details: [<u>CADET-Core Documentation</u>](https://cadet.github.io/v5.1.X/index.html)


## Solving the model

- Analytical solutions are generally unavailable for realistic chromatography models
- **Method of lines** is used to solve the PD(A)E that model chromatography:
    - Separate, sequential discretization of spatial and time coordinate
    - Spatial discretization yields an OD(A)E
    - ODAE solved by well-established, highly optimzed and freely available software
