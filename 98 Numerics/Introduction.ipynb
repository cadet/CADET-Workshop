## Solving the model

Generally no analytical solution available, we need some numerical (or statistical) method.
The so-called **method of lines** is most prominent for the PDAE at hand.
The method of lines considers spatial and temporal derivatives separately.
Usually, the spatial derivatives are discretized first, resulting in an ODAE system.
This is desireable because ODE and ODAE problems can be plugged into well-established, highly optimzed and freely availeable software.
The spatial discretization on the other hand requires the implementation of the custom model equations and a suitable spatial discretization method.
