# 0.1 Introduction to Types

Python (as most high-level programming languages) is based of a simple yet powerful philosophy: **Everything is an Object**.

Imagine you are organizing your home. Every object has:
- **A Type** (Book, Chair, Cup)
- **Properties** (Color, Size, Material)
- **Capabilities** (What can you do with it?)


Same principle applies in Python: Every object has a **type** with specific **properties** and **capabilities**.

## Why do types matter?

Types are like categories in real life. They help us:

1. **Safety**: We know what to expect
2. **Functionality**: Different types can do different things
3. **Abstraction**: We think in concepts, not in details

**Example from real life**:
- With a *hammer* you can hammer
- With a *screwdriver* you can screw
- With a *number* you can calculate
- With a *text* you can write

## For basic types in Python, we have:
- **Integers (`int`)**: Whole numbers, e.g., `1`, `42`, `-7`
- **Floating-point numbers (`float`)**: Decimal numbers, e.g., `3.14`, `-0.001`
- **Strings (`str`)**: Text, e.g., `"Hello, World!"`, `'Python is fun'`
- **Booleans (`bool`)**: True/False values, e.g., `True`, `False`



> ### **Why this matters for CADET-Process:**
- **Integers**: Number of components, discretization points, time steps
- **Floats**: Concentrations, flow rates, temperatures, column dimensions  
- **Strings**: Component names, unit operation types, file paths
- **Booleans**: Simulation flags, convergence status, configuration options

````python
# 1. Integers (whole numbers)
age = 25
year = 2024
temp = -5

print(f"Age: {age}")
print(f"Year: {year}")
print(f"Temperature: {temp}")
````

````python
# 2. Floats (decimal numbers)
pi = 3.14159
weight = 75.5
percentage = 0.85

print(f"Pi: {pi}")
print(f"Weight: {weight} kg")
print(f"Percentage: {percentage * 100}%")
````

````python
# 3. String (text)
name = "Anna"
city = 'Berlin'
message = """this is a
multiline text"""

print(f"Name: {name}")
print(f"City: {city}")
print(f"Message: {message}")
````

````python
# 4. Boolean (True/False)
is_student = True
has_license = False
is_adult = age >= 18  # This is a calculation!

print(f"Is student: {is_student}")
print(f"Has license: {has_license}")
print(f"Is adult: {is_adult}")
````

## Comparison Operators - Building Conditions

**Comparison operators** create `bool` values by comparing data:

**All Types:**
- `==` - equal to
- `!=` - not equal to

**Numbers (int, float):**
- `<` - less than
- `>` - greater than  
- `<=` - less than or equal
- `>=` - greater than or equal

**Strings:**
- Alphabetical comparison works with `<`, `>`, etc.
- Case sensitive: `"A" < "a"` is `True`

### **CADET-Process Connection:**
In CADET-Process, you'll frequently use comparisons to:
- Check if concentrations are above/below thresholds: `concentration > 0.5`
- Compare simulation results: `result1.solution_times == result2.solution_times`
- Validate parameters: `flow_rate >= 0` (flow rates can't be negative!)
- String comparisons for component names: `component_name == "Protein_A"`

````python
# Comparison operators with all basic types

# Numbers (int and float)
num1 = 10
num2 = 5.5
print("Number Comparisons:")
print(f"{num1} == {num2}: {num1 == num2}")
print(f"{num1} != {num2}: {num1 != num2}")
print(f"{num1} > {num2}: {num1 > num2}")
print(f"{num1} <= {num2}: {num1 <= num2}")

# Strings
str1 = "apple"
str2 = "banana"
print(f"\nString Comparisons:")
print(f"'{str1}' == '{str2}': {str1 == str2}")
print(f"'{str1}' != '{str2}': {str1 != str2}")
print(f"'{str1}' < '{str2}': {str1 < str2}")  # Alphabetical order
print(f"'{str1}' > '{str2}': {str1 > str2}")

# Case sensitivity
str3 = "Apple"
print(f"'{str1}' == '{str3}': {str1 == str3}")  # Case matters!
print(f"'{str1}' < '{str3}': {str1 < str3}")    # Capital letters come first

# Boolean comparisons
bool1 = True
bool2 = False
print(f"\nBoolean Comparisons:")
print(f"{bool1} == {bool2}: {bool1 == bool2}")
print(f"{bool1} != {bool2}: {bool1 != bool2}")
````

## Check types 

Sometimes we want to know what type a object has:
We have the built-in function 
- `type()` that returns the type of a given object.
- `isinstance()` that checks if an object is an instance of a specific class or type.

````python
# type() - function that returns the type of a object

theAnswer = 42
pi = 3.14
greeting = "Hello"
is_active = True

print(f"{theAnswer} has type: {type(theAnswer).__name__}")
print(f"{pi} has type: {type(pi).__name__}")
print(f"{greeting} has type: {type(greeting).__name__}")
print(f"{is_active} has type: {type(is_active).__name__}")

# __name__ to get the name of the type as a string, we will understand this later in 0.7 - Classes
````

````python
# isinstance() - Question: "Is this a specific type?"
theAnswer = 42
text = "Hello"

print(f"Is {theAnswer} a number (int) ? {isinstance(theAnswer, int)}")
print(f"Is {text} a number (int) ? {isinstance(text, int)}")
print(f"Is {text} a text (str) ? {isinstance(text, str)}")

# We also can check for multiple types at once
print(f"Is {theAnswer} a number (int or float) ? {isinstance(theAnswer, (int, float))}")
````

## Typecasting

The type of an object can be changed (if possible) using typecasting functions:
- `int()`: Converts to integer
- `float()`: Converts to floating-point number
- `str()`: Converts to string
- `bool()`: Converts to boolean

````python
# From text to number
text_number = "123"
real_number = int(text_number)
decimal_number = float(text_number)

print(f"Test    : '{text_number}' (Typ: {type(text_number)})")
print(f"As int: {real_number} (Typ: {type(real_number)})")
print(f"As float: {decimal_number} (Typ: {type(decimal_number)})")

# From numbers to text
age = 25
age_text = str(age)
print(f"\nNumber: {age} (Typ: {type(age)})")
print(f"As Text: '{age_text}' (Typ: {type(age_text)})")
````
