# 0.4 Containers - Organizing and Accessing Data

We've learned about individual types (`int`, `float`, `str`, `bool`), functions to organize code, and control flow to make decisions. 

Now we'll learn about **containers** - special types that can hold multiple values!

**The Big Idea:**
- **Individual values**: `42`, `"Python"`, `True` - single pieces of data
- **Containers**: `[1, 2, 3]`, `"Hello"` - collections that hold multiple pieces
- **Access patterns**: Use indexes `[0]` and slicing `[1:3]` to get specific parts

**Philosophy**: Containers organize data like folders organize files - they keep related information together and let us access it systematically!

**Why Do Containers Matter?**

1. **Real-World Data is Complex**: Most problems involve multiple related values - a list of students, characters in a name, items in a shopping cart
2. **Efficiency**: Instead of creating separate variables for each item, containers let us store and process collections efficiently
3. **Systematic Processing**: Containers work perfectly with loops - process all items with just a few lines of code
4. **Data Organization**: Like organizing books on shelves, containers keep related data together and accessible
5. **Scalability**: The same code works whether you have 3 items or 3,000 items in your container

**Example**: Instead of `student1 = "Alice"`, `student2 = "Bob"`, `student3 = "Charlie"`, we use `students = ["Alice", "Bob", "Charlie"]` and can easily add more students or process them all at once!

## Accessing List Items - Like an Address Book
Each object in a list can be accessed uniquely using its **index** (address).

Access by index is given with Square brackets `[]`. **Important note** (and very famous mistake): Python starts counting from 0!

But what if we want to access the last item in a list, without knowing its length? No problem! Python allows **negative indexing**: `-1` is the last item, `-2` the second last, and so on.

The square brackets can also be used to **change** an item at a specific index. Just assign a new value to that index!

### Slicing - Extract Parts of Lists 

**Slicing** is one of the most powerful features in Python! With the notation `[start:stop:step]` you can extract **sub-ranges** from lists (and later also strings).

**Syntax**: `list[start:stop:step]`
- **start**: Where to begin? (inclusive)
- **stop**: Where to stop? (exclusive!)
- **step**: Step size (optional, default: 1)

**Important**: The **stop** index is **not** included! `[0:3]` returns elements 0, 1, 2 - but **not** 3.

**Shortcuts**:
- `[:3]` = From beginning to index 3
- `[2:]` = From index 2 to end
- `[:]` = The complete list (copy!)
- `[::-1]` = Backwards (all elements reversed)




````python
fruits = ["apple", "banana", "orange", "grape"]

print(f"Original list: {fruits}")

# 1. CREATING LISTS with []
new_list = ["red", "green", "blue"]  # Create new list
empty_list = []                      # Create empty list
print(f"New colors: {new_list}")
print(f"Empty list: {empty_list}")
````

````python
# 2. ACCESSING ELEMENTS with [index]
first_fruit = fruits[0]  # Access first element
second_fruit = fruits[1] # Access second element
last_fruit = fruits[-1]  # Access last element using negative indexing
second_last_fruit = fruits[-2] # Access second to last element

print(f"First fruit [0]: {first_fruit}")
print(f"Second fruit [1]: {second_fruit}")
print(f"Last fruit [-1]: {last_fruit}")           # Negative indexing
print(f"Second to last [-2]: {second_last_fruit}")
````

````python
# 3. SLICING with [start:stop:step]
first_two = fruits[0:2]      # First two elements
from_index_one = fruits[1:]   # From index 1 to end
last_two = fruits[-2:]        # Last two elements
every_second = fruits[::2]    # Every second element
reversed_fruits = fruits[::-1] # Reversed list

print("\n3. SLICING (EXTRACTING PARTS):")
print(f"First two [0:2]: {first_two}")
print(f"From index 1 [1:]: {from_index_one}")
print(f"Last two [-2:]: {last_two}")
print(f"Every second [::2]: {every_second}")
print(f"Reversed [::-1]: {reversed_fruits}")
````

````python
# 4. MODIFYING ELEMENTS with [index] = new_value
fruits_copy = fruits.copy()  # Make a copy first
print(f"Before change: {fruits_copy}")
fruits_copy[1] = "mango"     # Change element at index 1
print(f"After change: {fruits_copy}")
````

````python
# 5. THE PATTERN: container[position_or_range]
text = "Python"  # Strings also work with []!
first_character = text[0]
last_three = text[-3:]
reversed_text = text[::-1]
print(f"String: '{text}'")
print(f"First character [0]: '{first_character}'")
print(f"Last 3 characters [-3:]: '{last_three}'")
print(f"Reversed [::-1]: '{reversed_text}'")
````

## Lists and For Loops - Processing Container Contents

Now that we understand containers and control flow, let's combine them! **For loops** are perfect for processing **all items** in a list systematically.

**Basic Pattern:**
```python
for item in my_list:
    # do something with each item
```

**Why This Works:**
- Lists are **sequences** - ordered collections of items
- For loops **iterate** through sequences item by item
- Each iteration, the loop variable gets the next item
- Automatic handling of container length and indexing

**Common Use Cases:**
- **Process all items**: Calculate totals, find patterns, transform data
- **Filter items**: Find items that match certain criteria  
- **Accumulate results**: Build up calculations or new containers
- **Validate data**: Check all items meet requirements

**Key Advantage**: No need to manage indexes manually - the loop handles container navigation automatically!

````python
# Lists and For Loops - Practical Examples

# 1. Basic iteration - process all items
numbers = [10, 25, 7, 42, 3, 18]
print(f"Numbers: {numbers}")

print("\nProcessing each number:")
for number in numbers:
    doubled = number * 2
    print(f"  {number} × 2 = {doubled}")
````

````python
# 2. Accumulation - calculate totals
total = 0
count = 0

for number in numbers:
    total += number
    count += 1
    print(f"  Added {number}, running total: {total}")

average = total / count
print(f"Final total: {total}, count: {count}, average: {average:.2f}")
````

````python
# 3. Filtering - find items matching criteria
large_numbers = []

for number in numbers:
    if number > 20:
        large_numbers.append(number)
        print(f"  {number} is large - added to collection")
    else:
        print(f"  {number} is small - skipped")

print(f"Large numbers found: {large_numbers}")

# 4. Processing mixed data types
mixed_data = [42, "hello", 3.14, True, "world", 100, False]
print(f"\nMixed data: {mixed_data}")

# Separate by type
integers = []
strings = []
floats = []
booleans = []

for item in mixed_data:
    item_type = type(item).__name__
    print(f"  Processing {item} ({item_type})")
    
    if isinstance(item, bool):  # Check bool first (bool is subtype of int)
        booleans.append(item)
    elif isinstance(item, int):
        integers.append(item)
    elif isinstance(item, float):
        floats.append(item)
    elif isinstance(item, str):
        strings.append(item)

print(f"Sorted by type:")
print(f"  Integers: {integers}")
print(f"  Strings: {strings}")
print(f"  Floats: {floats}")
print(f"  Booleans: {booleans}")
````

````python
# 6. Nested lists - lists containing lists
matrix = [
    [1, 2, 3],
    [4, 5, 6], 
    [7, 8, 9]
]

print(f"\nMatrix (nested lists): {matrix}")

print("Processing matrix rows:")
for row_index, row in enumerate(matrix):
    row_sum = 0
    print(f"  Row {row_index}: {row}")
    
    for number in row:
        row_sum += number
    
    print(f"    Row sum: {row_sum}")
````

## Advanced Function Parameters with Containers

Now that we understand containers and how to process them, let's explore how functions can work with **variable numbers of arguments** using containers!

**The Problem:**
Sometimes we want to write functions that can handle different numbers of inputs:
```python
add_two_numbers(1, 2)      # Works with 2 numbers
add_three_numbers(1, 2, 3) # Need different function for 3 numbers?
```

**The Solution: *args and **kwargs**
Python provides special syntax to collect arguments into containers:

### *args - Collect Positional Arguments into a Tuple-like Container
- `*args` collects **extra positional arguments** into a tuple (similar to a list)
- Useful when you don't know how many arguments will be passed
- The name `args` can be anything, the `*` is what matters


````python
# Advanced Function Parameters - *args and **kwargs with containers

# 1. *args - Variable positional arguments (creates tuple-like container)
def add_numbers(*args):
    """Add any number of arguments using container processing"""
    print(f"Received {len(args)} arguments: {args}")
    print(f"Type of args: {type(args).__name__}")
    
    # Process the args container like any other container
    total = 0
    for number in args:
        print(f"  Adding: {number}")
        total += number
    
    return total

# Test with different numbers of arguments
print("\n1. Testing *args:")
result1 = add_numbers(1, 2, 3)
print(f"Result: {result1}")

result2 = add_numbers(10, 20, 30, 40, 50)  
print(f"Result: {result2}")

result3 = add_numbers()  # No arguments - empty container
print(f"Result: {result3}")
````

````python
# 4. Unpacking containers into function arguments

# Have containers of data
number_list = [1, 2, 3, 4, 5]
person_info = {'name': 'Charlie', 'age': 30, 'city': 'Munich'}

print(f"Number list: {number_list}")
print(f"Person info: {person_info}")

# Unpack list with * (spreads list items as separate arguments)
sum_result = add_numbers(*number_list)  # Same as add_numbers(1, 2, 3, 4, 5)
print(f"Sum of unpacked list: {sum_result}")
````

## Exercise 2: 
**List Methods Mastery**

You are managing a **movie watchlist** for a film club. Start with this initial list:

```python
watchlist = ["Inception", "The Matrix", "Interstellar", "Blade Runner", "Ex Machina"]
```

Write functions that:

1. **Validate**: Checks if a given list is a watchlist
2. **Add Movies**: Adds movies to a given list
3. **Remove Movies**: Delets movies from a given list
4. **Analyze Your List**: Performs some statistics on the a given list

Create your own watchlist and test your functions.


````python
# Exercise 2: List Methods Mastery - Movie Watchlist Management

watchlist = ["Inception", "The Matrix", "Interstellar", "Blade Runner", "Ex Machina"]

def validate_watchlist(movie_list):
    """Checks if a given list is a proper watchlist"""
    if type(movie_list) != list:
        return False
    if len(movie_list) == 0:
        return False
    for item in movie_list:
        if type(item) != str:
            return False
    return True

def add_movies(movie_list, new_movie):
    """Adds a movie to the watchlist if it's not already there"""
    if type(new_movie) == str:
        if new_movie not in movie_list:
            movie_list.append(new_movie)
        else:
            print("Already in list")

def remove_movies(movie_list, movie_to_remove):
    """Removes a movie from the watchlist if it exists"""
    if movie_to_remove in movie_list:
        movie_list.remove(movie_to_remove)
    else:
        print("Not found")

def analyze_watchlist(movie_list):
    """Performs statistics on the watchlist"""
    if not validate_watchlist(movie_list):
        print("Invalid watchlist")
        return
    
    print(f"Total movies: {len(movie_list)}")
    
    # Find longest and shortest titles
    longest = movie_list[0]
    shortest = movie_list[0]
    
    for movie in movie_list:
        if len(movie) > len(longest):
            longest = movie
        if len(movie) < len(shortest):
            shortest = movie
    
    print(f"Longest title: {longest}")
    print(f"Shortest title: {shortest}")

# Test the functions
my_watchlist = ["Dune", "Avatar", "Terminator"]
print(f"My watchlist: {my_watchlist}")

# Test validation
print(f"Is valid? {validate_watchlist(my_watchlist)}")

# Test adding
add_movies(my_watchlist, "Blade Runner")
add_movies(my_watchlist, "Dune")  # Already exists
print(f"After adding: {my_watchlist}")

# Test removing
remove_movies(my_watchlist, "Avatar")
print(f"After removing: {my_watchlist}")

# Test analysis
analyze_watchlist(my_watchlist)

# CADET example with similar patterns
components = ["Salt", "Buffer", "Protein_A"]

def validate_components(comp_list):
    """Check if component list is valid for CADET"""
    if type(comp_list) != list:
        return False
    if len(comp_list) == 0:
        return False
    for comp in comp_list:
        if type(comp) != str:
            return False
    return True

def add_component(comp_list, new_comp):
    """Add component to CADET process"""
    if new_comp not in comp_list:
        comp_list.append(new_comp)

print(f"Components: {components}")
add_component(components, "Protein_B")
print(f"Updated: {components}")
print(f"Valid? {validate_components(components)}")
````
