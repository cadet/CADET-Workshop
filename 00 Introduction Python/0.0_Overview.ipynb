# **Welcome to the Python Introduction for the CADET Workshop**

## **Intention & Goals**
    
This Python tutorial provides a **focused introduction** tailored specifically for participants of the **CADET Workshop**.
This tutorial is designed for everybody who has no or very little prior programming experience.
It is **not** a full-scale programming course or a comprehensive Python guide.

## **Tutorial Structure & Learning Path**

The tutorial is organized into **8 progressive Jupyter notebooks**, taking you from beginner level to **CADET-ready**.

### **Phase 1: Python Fundamentals**

**Goal:** Understand Python’s basic philosophy and fundamental building blocks.

| Notebook | Topic | Key Concepts | Duration |
|-----------|--------|---------------|-----------|
| **0.1** | **Types & Objects** | `int`, `float`, `str`, `bool` – Everything is an object | ~20 min |
| **0.2** | **Functions** | Code organization, reusability | ~20 min |
| **0.3** | **Control Flow** | `if/else`, loops, making decisions | ~20 min |

**Learning Outcome:** You can write basic Python programs that process data and make decisions.

---

### **Phase 2: Data Organization**

**Goal:** Learn to manage and structure data effectively for scientific computing.

| Notebook | Topic | Key Concepts | Duration |
|-----------|--------|---------------|-----------|
| **0.4** | **Containers (Lists)** | Collections, indexing, iteration | ~30 min |
| **0.5** | **Methods** | Built-in functionality, dot notation | ~15 min |
| **0.6** | **Dictionaries** | Key-value pairs, structured data, `**kwargs` | ~15 min |

**Learning Outcome:** You can organize and manipulate structured data.

---

### **Phase 3: Object-Oriented Programming**

**Goal:** Understand object-oriented programming and software architecture principles.

| Notebook | Topic | Key Concepts | Duration |
|-----------|--------|---------------|-----------|
| **0.7** | **Classes** | Custom types, inheritance, *CADET Component Challenge* | ~75 min |
| **0.8** | **Libraries & Environment** | Scientific ecosystem, package management | ~15 min |

**Learning Outcome:** You understand the importance of libraries and virtual environments in Python.

---

## **Progressive Complexity**

Each notebook builds upon the previous one:

**Types → Functions → Control Flow → Containers → Methods → Dictionaries → Classes → Libraries**

### **What This Tutorial IS**

- Starts from the very beginning — designed for participants **new to programming and Python**  
- Explains the key programming concepts on which **CADET-Process** is built  
- Emphasizes **hands-on examples** and **practical exercises**

### **What This Tutorial IS NOT**

- Does **not** cover all aspects of Python
- Avoids deep theoretical discussions
- Excludes complex algorithms or data structures and handling

---

### **Self-Assessment Questions**

After completing the tutorial, you should be able to answer:

1. Can I create and manipulate variables, lists, and dictionaries?  
2. Can I write reusable functions with parameters?  
3. Can I implement control flow with conditions and loops?  
4. Do I understand how CADET-Process components work as classes?
5. Can I import and use scientific libraries?
