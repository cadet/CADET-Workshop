# 0.5 Methods - Built-in Functions for Every Type

We've learned about types, functions, control flow, and containers. Now we'll discover that **every type comes with built-in functions called methods**!

**The Idea:**
- **Functions we create**: `def my_function(data):` - we write the logic ourselves
- **Methods**: `data.method()` - the logic is already written and optimized!
- **The Magic**: Methods belong to objects and automatically know their type

**Philosophy**: Methods are like specialized tools that each type provides - a list knows how to add items, a string knows how to change case, numbers know their properties!

## Dot Notation - The Universal Python Pattern

**Dot notation** is how we connect objects with their built-in capabilities!

**The Pattern:**
```python
object.method()      # Call a method
```

**What happens:**
1. `object` - Python identifies the type of this object
2. `.` - The dot says "I want to use a capability of this object"
3. `method()` - Python finds and executes the method for that specific type

**The Power:** Each type has different methods that make sense for that type!



````python
# Demonstrating dot notation with different types

# Numbers have methods too!
number = 42
print(f"Number: {number}")
print(f"Binary representation: {number.bit_length()} bits needed")
````

````python
# Strings have many methods
text = "Python Programming"
print(f"\nString: '{text}'")
print(f"Uppercase: '{text.upper()}'")
print(f"Lowercase: '{text.lower()}'")
print(f"Word count: {len(text.split())}")
````

````python
# Lists have methods for manipulation  
my_list = [1, 2, 3]
print(f"\nList: {my_list}")
my_list.append(4)  # Method call - modifies the list
print(f"After append(4): {my_list}")
````

````python
# Booleans are objects too
flag = True
print(f"\nBoolean: {flag}")
print(f"Boolean bit length: {flag.bit_length()}")  # True = 1, False = 0

print("\nKey insight: The DOT connects objects with their capabilities!")
print("Each type knows what it can do and provides methods accordingly.")
````

# Example: List Methods - Containers with Built-in Intelligence

Lists come with powerful methods that let you modify and analyze them efficiently!

**Key List Methods:**
- **Adding**: `.append()`, `.extend()`, `.insert()`
- **Removing**: `.remove()`, `.pop()`, `.clear()`
- **Organizing**: `.sort()`, `.reverse()`
- **Finding**: `.index()`, `.count()`
- **Copying**: `.copy()`

**Important Concept:** List methods **modify the original list** (they're mutable)!

````python
# List methods in action - the power of built-in functions!
hobbies = ["reading", "sports", "gaming"]
print(f"Starting hobbies: {hobbies}")

# .append() - Adds ONE element to the end
hobbies.append("music")
print(f"After append('music'): {hobbies}")
````

````python
# .extend() - Adds MULTIPLE elements from another list
hobbies.extend(["cooking", "traveling"])
print(f"After extend(['cooking', 'traveling']): {hobbies}")

# .insert() - Inserts at a specific position
hobbies.insert(1, "programming")  # Insert at index 1
print(f"After insert(1, 'programming'): {hobbies}")
````

````python
# .remove() - Removes the first occurrence of a value
hobbies.remove("sports")
print(f"After remove('sports'): {hobbies}")

# .pop() - Removes and returns an element (default: last)
removed_hobby = hobbies.pop()  # Remove last
print(f"Popped: '{removed_hobby}', remaining: {hobbies}")

removed_first = hobbies.pop(0)  # Remove first (index 0)
print(f"Popped first: '{removed_first}', remaining: {hobbies}")
````

````python
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"Original numbers: {numbers}")

# .sort() - Sorts the list in place
numbers.sort()
print(f"After sort(): {numbers}")

# .reverse() - Reverses the order
numbers.reverse()
print(f"After reverse(): {numbers}")
````

````python
letters = ['a', 'b', 'c', 'b', 'd', 'b']
print(f"Letters: {letters}")

# .count() - Count occurrences
b_count = letters.count('b')
print(f"Letter 'b' appears {b_count} times")
````

````python
# .index() - Find first occurrence position
b_position = letters.index('b')
print(f"First 'b' is at index {b_position}")
````

````python
original = [1, 2, 3, 4, 5]
print(f"Original: {original}")

# .copy() - Creates a new copy
backup = original.copy()
backup.append(6)
print(f"Original after backup modified: {original}")
print(f"Backup: {backup}")
````

````python
# .clear() - Empties the list
test_list = [1, 2, 3]
print(f"Before clear(): {test_list}")
test_list.clear()
print(f"After clear(): {test_list}")
````

# Methods vs Functions - Understanding the Difference

**This is crucial for understanding Python!**

## Functions (What we create)
```python
def my_function(data):
    return data.upper()

result = my_function("hello")  # We pass data TO the function
```

## Methods (What objects provide)  
```python
text = "hello"
result = text.upper()  # The object knows how to work on itself
```

## Key Differences

| Aspect | Functions | Methods |
|--------|-----------|---------|
| **Who owns it** | Nobody (standalone) | The object/type |
| **How to call** | `function(object)` | `object.method()` |
| **Who writes it** | You | Python/library authors |
| **Knowledge** | Doesn't know object type | Knows its object type |

## Why Methods Are Powerful

1. **Type-specific behavior**: A list's `.append()` knows it's working with a list
2. **Optimized**: Written by experts for maximum efficiency  
3. **Consistent**: Same patterns across all Python objects
4. **Self-documenting**: `text.upper()` clearly shows what's happening

## Exercise 1: Method vs Function Design

Compare and contrast different approaches to the same problem:

**Your Tasks:**
- Write a function that removes duplicates from a list and returns a sorted list
- Solve the same problem using both custom functions AND built-in methods
- Compare performance and readability
- Understand when to use functions vs methods


````python
# Exercise 1: Method vs Function Design

# Original data with duplicates
original_data = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5, 8, 9]
print(f"Original data: {original_data}")

# Approach 1: Using custom function
def remove_duplicates_and_sort_function(data):
    """Custom function approach - more explicit control"""
    # Create a new list to store unique items
    unique_items = []
    
    # Check each item and add if not already present
    for item in data:
        if item not in unique_items:
            unique_items.append(item)
    
    # Sort the list
    sorted_items = sorted(unique_items)
    return sorted_items

# Approach 2: Using built-in methods (more Pythonic)
def remove_duplicates_and_sort_methods(data):
    """Using built-in methods - cleaner and more efficient"""
    # Convert to set (removes duplicates), then back to list, then sort
    unique_sorted = sorted(list(set(data)))
    return unique_sorted

# Approach 3: One-liner using methods
def remove_duplicates_and_sort_oneliner(data):
    """Most concise approach"""
    return sorted(set(data))

# Test all approaches
print(f"\nApproach 1 (custom function): {remove_duplicates_and_sort_function(original_data)}")
print(f"Approach 2 (built-in methods): {remove_duplicates_and_sort_methods(original_data)}")
print(f"Approach 3 (one-liner): {remove_duplicates_and_sort_oneliner(original_data)}")

# Performance comparison (simple timing)
import time

# Test with larger dataset
large_data = original_data * 1000

# Time approach 1
start = time.time()
result1 = remove_duplicates_and_sort_function(large_data)
time1 = time.time() - start

# Time approach 2
start = time.time()
result2 = remove_duplicates_and_sort_methods(large_data)
time2 = time.time() - start

print(f"\nPerformance comparison (1000x data):")
print(f"Custom function: {time1:.6f} seconds")
print(f"Built-in methods: {time2:.6f} seconds")
print(f"Built-in methods are {time1/time2:.1f}x faster!")
````
