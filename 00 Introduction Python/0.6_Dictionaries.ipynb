## The Philosophy: From Sequential to Named Data

Lists organize data by **position**: `["Anna", "Ben", "Clara"]`

Dictionaries organize data by **meaning**: `{"name": "Anna", "age": 28, "city": "Berlin"}`

This makes code more **readable** and **maintainable**!

# Dictionaries - Structured Data with Names

Lists are great for **sequences**: `["Anna", "Ben", "Clara"]`

But what if we want to **know more about Anna**? Age, city, profession?

**Dictionaries solve the problem**: They connect **keys** (names) with **values** (the data we already know - int, str, bool, lists!).

###  **CADET-Process Connection:**
Dictionaries are perfect for CADET-Process configuration:
- **Component properties**: `{"name": "Protein_A", "molar_mass": 15000, "charge": -2}`
- **Process parameters**: `{"flow_rate": 1.5, "temperature": 25, "pH": 7.4}`
- **Column specifications**: `{"length": 0.1, "diameter": 0.02, "porosity": 0.37}`
- **Binding model data**: `{"ka": 1.2e4, "kd": 0.05, "nu": 2.8}`

````python
# Creating dictionaries - different methods
person = {
    "name": "Anna",           # str as value
    "age": 28,                # int as value  
    "city": "Berlin",         # str as value
    "is_student": False,      # bool as value
    "hobbies": ["Reading", "Sports", "Music"] # list as value!
}

print(f"Person: {person}")
print(f"Type: {type(person)}")
````

````python

# Access values via keys
person_name = person["name"]
person_age = person["age"]
person_first_hoppy = person['hobbies'][0]

print(f"\nName: {person['name']}")
print(f"Age: {person['age']}")
print(f"First hobby: {person['hobbies'][0]}")
````

````python
# Alternative syntax
city = person.get('city')
job = person.get('job')

print(f"City: {city}")
print(f"Job: {job}")
````

## Dictionary Operations - Managing Data Dynamically

Dictionaries are **mutable** - we can add, change, and remove keys:

````python
# Start with a simple dictionary
student = {
    "name": "Max",
    "age": 22
}
print(f"Start: {student}")
````

````python
# Add new keys
student["major"] = "Computer Science"
student["semester"] = 3
print(f"After addition: {student}")
````

````python
# Change values
student["age"] = 23  # Birthday!
print(f"After birthday: {student}")
````

````python
# Remove keys
removed_value = student.pop("semester")
print(f"Removed: {removed_value}")
print(f"After removal: {student}")
````

## Dictionary Methods 

Remember list methods like `.append()` and string methods like `.upper()`? Dictionaries have their own methods too!

````python
configuration = {
    "language": "english",
    "theme": "dark",
    "notifications": True
}
print(f"Configuration: {configuration}")


keys = configuration.keys() # .keys() - All keys
values = configuration.values() # .values() - All values 
items = configuration.items() # .items() - Key-value pairs

print(f"\nKeys: {list(keys)}")
print(f"Values: {list(configuration.values())}")
print(f"Pairs: {list(configuration.items())}")
````

````python
# .update() - Change/add multiple values at once
configuration.update({
    "theme": "light",
    "font_size": 14,
    "auto_save": True
})
print(f"\nAfter update: {configuration}")
````

````python
# .copy() - Create copy (like with lists!)
backup = configuration.copy()
print(f"Backup created: {len(backup)} entries")
````

## Nested Structures

The brilliant thing: Dictionaries can contain **anything** as values - even other dictionaries, list and functions

````python
# Complex data structure

# A dictionary can even contain functions!
def calculate_bonus(salary, performance_rating):
    """Calculate bonus based on salary and performance (1-5 scale)"""
    bonus_percentage = performance_rating * 0.05  # 5% per rating point
    return salary * bonus_percentage

# Adding a function to our dictionary
company = {
    "name": "Tech Solutions Ltd",
    "founded": 2020,
    "employees": [
        {
            "name": "Anna Schmidt",
            "position": "Developer",
            "salary": 55000,
            "skills": ["Python", "JavaScript", "SQL"]
        },
        {
            "name": "Ben Weber", 
            "position": "Designer",
            "salary": 48000,
            "skills": ["Photoshop", "Figma", "CSS"]
        }
    ],
    "locations": {
        "headquarters": "Berlin",
        "branch": "Munich",
        "remote_percentage": 0.6
    },
    "calculate_bonus": calculate_bonus
}
````

````python
# Navigate through the structure
company_name = company["name"]
first_employee_name = company["employees"][0]["name"]
first_employee_first_skill = company["employees"][0]["skills"][0]
headquarters_location = company["locations"]["headquarters"]

print(f"Company: {company_name}")
print(f"First employee: {first_employee_name}")
print(f"Their first skill: {first_employee_first_skill}")
print(f"Headquarters: {headquarters_location}")
````

````python
# Using the function stored in the dictionary
anna_bonus = company['calculate_bonus'](company['employees'][0]['salary'], 4)
ben_bonus = company['calculate_bonus'](company['employees'][1]['salary'], 3)

print(f"\nBonus calculations:")
print(f"Anna's bonus (rating 4): ${anna_bonus:,.0f}")
print(f"Ben's bonus (rating 3): ${ben_bonus:,.0f}")
````

````python
# Calculate statistics
total_salary = sum(emp['salary'] for emp in company['employees'])
average = total_salary / len(company['employees'])
print(f"\nAverage salary: ${average:,.0f}")
````

## Dictionaries and **kwargs - The Connection Revealed! 

Remember `*args` from the containers notebook? Now that you understand dictionaries, let's explore `**kwargs`!

**The Pattern:**
- `*args` collects **positional arguments** into a **tuple** (like a list)
- `**kwargs` collects **keyword arguments** into a **dictionary**!

**Why This Works:**
- Dictionary keys are **strings** (perfect for parameter names)
- Dictionary values can be **any type** (perfect for parameter values)
- Dictionaries organize data by **name** (perfect for named parameters)

**The Magic**: `**kwargs` creates a dictionary where:
- **Keys** = parameter names (as strings)
- **Values** = parameter values (any type)

This makes functions incredibly flexible and readable!

````python
# **kwargs - creating dictionaries from function parameters

def create_user_profile(**kwargs):
    """
    Create a user profile dictionary from keyword arguments
    **kwargs automatically collects all keyword arguments into a dictionary!
    """
    print(f"Received kwargs: {kwargs}")
    print(f"Type of kwargs: {type(kwargs)}")
    
    # kwargs IS a dictionary - we can use all dictionary operations!
    profile = kwargs.copy()  # Start with provided data
    
    # Add some default values if not provided
    if 'status' not in profile:
        profile['status'] = 'active'
    
    if 'created_date' not in profile:
        from datetime import datetime
        profile['created_date'] = datetime.now().strftime("%Y-%m-%d")
    
    return profile

# Test the function - see how kwargs becomes a dictionary

# Call with different keyword arguments
user1 = create_user_profile(name="Anna", age=28, city="Berlin", profession="Developer")
print(f"User 1: {user1}")

user2 = create_user_profile(name="Ben", age=32, hobby="Photography", status="premium")
print(f"User 2: {user2}")

# Empty call
user3 = create_user_profile()
print(f"User 3: {user3}")
````

````python
# We can unpack a dictionary into keyword arguments
existing_data = {
    "name": "Clara",
    "age": 25,
    "city": "Munich",
    "skills": ["Python", "Data Science"]
}

print(f"Existing data: {existing_data}")

# Unpack dictionary as keyword arguments using **
user4 = create_user_profile(**existing_data)
print(f"User 4 (from unpacked dict): {user4}")

# Combine regular keywords with unpacked dictionary
user5 = create_user_profile(profession="Scientist", **existing_data, status="researcher")
print(f"User 5 (mixed): {user5}")
````

### Exercise 2: Student Grade Manager
Create a function `manage_grades(**kwargs)` that:
- Accepts student information as keyword arguments
- Creates a grade dictionary with subjects and scores
- Calculates and adds the average grade
- Returns the complete student profile

**Test with:**
- `manage_grades(name="Lisa", math=85, english=92, science=78)`
- `manage_grades(name="Tom", history=88, art=95, math=82, physics=79)`


````python
# Exercise 2: Student Grade Manager

def manage_grades(**kwargs):
    """
    Accepts student information as keyword arguments.
    Creates a grade dictionary with subjects and scores.
    Calculates and adds the average grade.
    Returns the complete student profile.
    
    Parameters:
    **kwargs: Student information including name and subject grades
    
    Returns:
    dict: Complete student profile with grades and average
    """
    print(f"Received kwargs: {kwargs}")
    
    # Start with a copy of all provided data
    student_profile = kwargs.copy()
    
    # Extract the name (required field)
    if 'name' not in student_profile:
        student_profile['name'] = "Unknown Student"
    
    # Separate name from grades
    name = student_profile['name']
    grades = {}
    
    # Collect all subjects and their grades
    for key, value in kwargs.items():
        if key != 'name':  # Everything except name is a subject grade
            if isinstance(value, (int, float)):  # Make sure it's a number
                grades[key] = value
    
    # Calculate average grade
    if grades:
        total_points = sum(grades.values())
        average = total_points / len(grades)
    else:
        average = 0
    
    # Create the complete profile
    complete_profile = {
        'name': name,
        'grades': grades,
        'average': round(average, 2),
        'total_subjects': len(grades)
    }
    
    return complete_profile

# Test the function
print("Student Grade Manager")
print("=" * 50)

# Test case 1
student1 = manage_grades(name="Lisa", math=85, english=92, science=78)
print(f"\nStudent 1: {student1['name']}")
print(f"Grades: {student1['grades']}")
print(f"Average: {student1['average']}")
print(f"Total subjects: {student1['total_subjects']}")

# Test case 2
student2 = manage_grades(name="Tom", history=88, art=95, math=82, physics=79)
print(f"\nStudent 2: {student2['name']}")
print(f"Grades: {student2['grades']}")
print(f"Average: {student2['average']}")
print(f"Total subjects: {student2['total_subjects']}")

# Test case 3 - student without name
student3 = manage_grades(chemistry=90, biology=87)
print(f"\nStudent 3: {student3['name']}")
print(f"Grades: {student3['grades']}")
print(f"Average: {student3['average']}")

# Additional functionality - Grade analysis
def analyze_student_performance(student_profile):
    """Analyze student performance based on their profile"""
    name = student_profile['name']
    average = student_profile['average']
    grades = student_profile['grades']
    
    print(f"\nPerformance Analysis for {name}:")
    print(f"Overall Average: {average}")
    
    # Find best and worst subjects
    if grades:
        best_subject = max(grades, key=grades.get)
        worst_subject = min(grades, key=grades.get)
        
        print(f"Best subject: {best_subject} ({grades[best_subject]})")
        print(f"Needs improvement: {worst_subject} ({grades[worst_subject]})")
        
        # Performance category
        if average >= 90:
            category = "Excellent"
        elif average >= 80:
            category = "Good"
        elif average >= 70:
            category = "Satisfactory"
        else:
            category = "Needs Improvement"
        
        print(f"Performance Category: {category}")

# Analyze both students
analyze_student_performance(student1)
analyze_student_performance(student2)
````

### Exercise 3: Data Analysis Challenge
Given this nested company data structure, write code to:

```python
company = {
    "employees": [
        {"name": "Alice", "department": "Engineering", "salary": 70000, "skills": ["Python", "C++", "SQL"]},
        {"name": "Bob", "department": "Marketing", "salary": 50000, "skills": ["SEO", "Content Creation"]},
        {"name": "Charlie", "department": "Engineering", "salary": 80000, "skills": ["Java", "Python"]},
        {"name": "Diana", "department": "HR", "salary": 60000, "skills": ["Recruitment", "Employee Relations"]},
        {"name": "Eve", "department": "Engineering", "salary": 75000, "skills": ["Python", "SQL", "Machine Learning"]},
    ]
}
```
1. Find the employee with the highest salary
2. List all unique skills across all employees
3. Calculate total salary cost by department
4. Find employees who have both "Python" and "SQL" skills


````python
# Exercise 3: Data Analysis Challenge

company = {
    "employees": [
        {"name": "Alice", "department": "Engineering", "salary": 70000, "skills": ["Python", "C++", "SQL"]},
        {"name": "Bob", "department": "Marketing", "salary": 50000, "skills": ["SEO", "Content Creation"]},
        {"name": "Charlie", "department": "Engineering", "salary": 80000, "skills": ["Java", "Python"]},
        {"name": "Diana", "department": "HR", "salary": 60000, "skills": ["Recruitment", "Employee Relations"]},
        {"name": "Eve", "department": "Engineering", "salary": 75000, "skills": ["Python", "SQL", "Machine Learning"]},
    ]
}

print("Company Data Analysis Challenge")
print("=" * 50)

# 1. Find the employee with the highest salary
print("1. Employee with the highest salary:")
highest_paid_employee = None
highest_salary = 0

for employee in company["employees"]:
    if employee["salary"] > highest_salary:
        highest_salary = employee["salary"]
        highest_paid_employee = employee

print(f"Name: {highest_paid_employee['name']}")
print(f"Department: {highest_paid_employee['department']}")
print(f"Salary: ${highest_paid_employee['salary']:,}")

# 2. List all unique skills across all employees
print("\n2. All unique skills across employees:")
all_skills = []

for employee in company["employees"]:
    for skill in employee["skills"]:
        if skill not in all_skills:
            all_skills.append(skill)

print(f"Total unique skills: {len(all_skills)}")
for skill in all_skills:
    print(f"  - {skill}")

# 3. Calculate total salary cost by department
print("\n3. Total salary cost by department:")
department_costs = {}

for employee in company["employees"]:
    department = employee["department"]
    salary = employee["salary"]
    
    if department in department_costs:
        department_costs[department] += salary
    else:
        department_costs[department] = salary

for department, total_cost in department_costs.items():
    print(f"{department}: ${total_cost:,}")

# Calculate total company salary cost
total_company_cost = sum(department_costs.values())
print(f"\nTotal company salary cost: ${total_company_cost:,}")

# 4. Find employees who have both "Python" and "SQL" skills
print("\n4. Employees with both Python and SQL skills:")
python_sql_employees = []

for employee in company["employees"]:
    employee_skills = employee["skills"]
    has_python = "Python" in employee_skills
    has_sql = "SQL" in employee_skills
    
    if has_python and has_sql:
        python_sql_employees.append(employee)

print(f"Found {len(python_sql_employees)} employees with both Python and SQL skills:")
for employee in python_sql_employees:
    print(f"  - {employee['name']} ({employee['department']})")
    print(f"    Skills: {', '.join(employee['skills'])}")

# Additional analysis: Department statistics
print("\n" + "="*50)
print("Additional Analysis: Department Statistics")

department_stats = {}

for employee in company["employees"]:
    department = employee["department"]
    
    if department not in department_stats:
        department_stats[department] = {
            "employee_count": 0,
            "total_salary": 0,
            "employees": []
        }
    
    department_stats[department]["employee_count"] += 1
    department_stats[department]["total_salary"] += employee["salary"]
    department_stats[department]["employees"].append(employee["name"])

for department, stats in department_stats.items():
    average_salary = stats["total_salary"] / stats["employee_count"]
    print(f"\n{department} Department:")
    print(f"  Employees: {stats['employee_count']}")
    print(f"  Total Cost: ${stats['total_salary']:,}")
    print(f"  Average Salary: ${average_salary:,.0f}")
    print(f"  Team Members: {', '.join(stats['employees'])}")
````
