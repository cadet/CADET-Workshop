## Workshop structure

| Chapter                                  | Focus                          | Topics                                                                                      |
| ---------------------------------------- | ------------------------------ | ------------------------------------------------------------------------------------------- |
| 02 CADET Fundamentals                    | Building and simulating models | first model, model templates, inlet profiles, residence time distributions, binding         |
| 03 Process Simulation                    | Chromatographic processes      | simple and advanced chromatography                                                          |
| 04 Characterization and Optimization     | Using models                   | optimization essentials, parameter fitting and characterization, fractionation, process optimization |

The first lessons build models in interactive widgets and move step by step to code. From **02.5 Binding** onwards, we work in CADET-Process code; **Show code** from earlier lessons and the [GUI Workbench](../../99%20Misc/99.2%20GUI%20Workbench/99.2a_gui_workbench_lesson.ipynb) remain available as a fallback.

The chapters build on each other, but each topic can also be worked through on its own. [99.3 Reactions](../../99%20Misc/99.3%20Reactions/99.3a_reactions_lesson.ipynb) is an optional excursion. Detailed optimization examples are additional material in **99.4–99.5**, and [99.6 Common Problems](../../99%20Misc/99.6%20Common%20Problems/99.6a_common_problems_lesson.ipynb) collects typical modeling problems and how to check them.

[01.4 Contributing](../01.4%20Contributing/01.4a_contributing_lesson.ipynb) explains how to ask questions, report problems and propose changes to the CADET projects. The [01.5 Glossary](../01.5%20Glossary/01.5a_glossary_lesson.ipynb) defines the terms used throughout the workshop.

## Lessons and exercises

| | Lesson | Exercises |
|---|---|---|
| **You** | follow the presentation | write the code yourself, alone or in pairs |
| **Code** | shown and typed live by the presenter | empty answer cells for you to fill |
| **Solutions** | complete on the `solution` branch | complete on the `solution` branch |

During lessons, **please do not type along**. Everything typed live is filled in on the `solution` branch, so there is nothing to copy down. Your own coding starts in the exercises, which practise the steps shown in the lesson.

Stuck or behind? Each topic starts from a ready-made state, so you can always continue with the next lesson.

## Teaching and solution branch

The workshop lives in one repository, [CADET-Workshop](https://github.com/cadet/CADET-Workshop), with two branches:

| Branch     | Contains                                         | Use it                       |
| ---------- | ------------------------------------------------ | ---------------------------- |
| `teaching` | lessons and exercises, solution cells empty      | during the workshop          |
| `solution` | the same notebooks, all solution cells filled in | to check your work afterwards |

Both are generated from the same source, so cells line up one to one.

## Working in JupyterLab

| Name | What it is |
|---|---|
| **Jupyter notebook** | a document (`.ipynb`) of text and code cells; code runs in a Python **kernel** and its output appears below the cell |
| **JupyterLab** | the program in the browser in which notebooks are opened, edited and run, with a file browser and terminal |
| **JupyterHub** | a server that starts a JupyterLab for each user, so nothing has to be installed; the workshop runs on [jupyter.cadet-web.de](https://jupyter.cadet-web.de) |

- **Shift+Enter** runs a cell; run cells **from top to bottom**.
- Start over: *Kernel → Restart Kernel and Run All Cells*.

## Finding your way in the files

| Where | What |
|---|---|
| **File browser** (folder icon in the left side bar) | the workshop folders: chapter → topic → notebooks |
| `CADET-Workshop-2026-teaching` (JupyterHub) | the notebooks to work in |
| `CADET-Workshop-2026-solution` (JupyterHub) | the same notebooks with all answers |
| `02.1a_first_model_lesson.ipynb` | a lesson, as presented |
| `02.1b_first_model_exercises.ipynb` | the exercises to that lesson |

- A **double-click** on an `.ipynb` file opens it in a new tab.
- Dragging a tab to the right edge shows two notebooks side by side, e.g. a lesson and its exercises.
- The links at the end of each notebook open its exercises and the next lesson directly.
- `[*]` next to a cell means it is still running; `[3]` means it was the third cell run.

Before the first lesson, we open **02.1a First Model** together in the teaching folder and run its first cell.
