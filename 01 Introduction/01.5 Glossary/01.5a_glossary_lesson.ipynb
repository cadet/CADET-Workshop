## Terms that are easy to confuse

| Terms | Difference |
|---|---|
| **Data** vs. **model** | Data are measured values, e.g. a detector signal over time. A model predicts such values from equations and parameters. In fitting, the data are the **reference** that the simulated curve is compared with. |
| **Model parameter** vs. **variable** | A model parameter is a value in the model equations, e.g. `total_porosity` or `capacity`. A variable is a value an optimizer may change. Every variable writes into a parameter or operating setting; most parameters are never variables. |
| **Objective** vs. **constraint** | The objective is the score to minimize or maximize, e.g. the fitting error or productivity. A constraint is a requirement a candidate must meet, e.g. purity ≥ 95%; it decides which candidates are acceptable, not which is best. |
| **Single-** vs. **multi-objective** | One objective gives one best candidate. Several competing objectives, e.g. productivity and purity, give a **Pareto front** of trade-offs instead. |
| **Convergence** (three meanings) | **Optimizer:** the score stops improving between iterations; reaching an iteration limit is not convergence. **Cyclic stationarity:** successive cycles of a process give the same chromatogram. **Grid:** refining the discretization no longer changes the result. |
| **Total**, **bed** and **particle porosity** | Bed porosity: liquid volume between the particles per column volume. Particle porosity: pore volume per particle volume. Total porosity: all liquid volume per column volume; models without pores use only this one. |
| **Residence time** vs. **retention time** | Residence time: how long liquid spends in a unit, volume / flow rate. Retention time: when a component leaves the column; binding makes it longer than the residence time. |

## Building a model

| Term | Meaning | First used in |
|---|---|---|
| **Axial dispersion** | Spreading of a front along the flow direction, `axial_dispersion` in m²/s | 02.1, 02.4 |
| **Binding model** | Equations for the exchange between liquid and bound phase, e.g. `Linear`, `Langmuir`, `StericMassAction` (SMA) | 02.5, 03.1 |
| **Component** | A substance whose concentration is modeled, e.g. tracer, salt, a protein; listed in a `ComponentSystem` | 02.1 |
| **Cycle time** | Duration of one process cycle, `cycle_time` in seconds | 02.1 |
| **Discretization** | Splitting a unit into a finite number of cells for the numerical solution, e.g. `discretization.ncol` | 02.4 |
| **Event** | A change of a parameter at a given time, e.g. switching the inlet concentration, added with `add_event()` | 02.3 |
| **Flow sheet** | The unit operations and their connections, `FlowSheet` | 02.1 |
| **Gradient** | A gradual change of an inlet concentration over time, e.g. linear salt gradient elution (LWE) | 02.2, 02.3 |
| **Initial condition** | Concentrations at the start of a simulation, e.g. `c` and `q` | 02.1 |
| **Isotherm** | The equilibrium relation between liquid and bound concentration | 02.5 |
| **Kinetic binding** | Binding with finite rates (`is_kinetic = True`), in contrast to instant equilibrium | 02.5 |
| **Process** | A flow sheet with its operation over time: cycle time and events, `Process` | 02.1 |
| **Process template** | A prepared process such as Breakthrough, Pulse Injection or LWE | 02.2 |
| **Simulation** | Solving the model equations of a process over time, with `Cadet().simulate(process)` | 02.1 |
| **State** | A modeled concentration: `c` in the liquid, `cp` in the particle pores, `q` bound | 02.1, 03.1 |
| **Unit operation** | One part of the flow sheet, e.g. `Inlet`, `Cstr`, `TubularReactor`, column, `Outlet` | 02.1 |

## Experiments and characterization

| Term | Meaning | First used in |
|---|---|---|
| **Breakthrough curve** | Outlet concentration over time during a constant feed | 02.1 |
| **Characterization** | Determining model parameters from experiments, from the system periphery to the column and binding | 02.4, 04.2 |
| **Comparator** | Compares a simulated curve with reference data using one or more metrics | 04.2 |
| **Extra-column volume** | Liquid volume outside the column, in mixer and tubing; also **system volume**. Divided by the flow rate it gives the **system delay** | 02.4 |
| **Fitting error** | A score for the difference between simulated and measured curves, e.g. sum of squared errors (SSE) or root mean squared error (RMSE) | 04.1, 04.2 |
| **Pulse** | A short injection of a component, followed by component-free liquid | 02.2, 02.4 |
| **Reference** | Measured (or synthetic) data used for comparison, e.g. `ReferenceIO` | 04.1, 04.2 |
| **Tracer** | A non-binding component used to measure transport, e.g. Acetone (enters the pores) or Dextran (does not) | 02.1 |

|---|---|---|

## Optimization

| Term | Meaning | First used in |
|---|---|---|
| **Bound** | The allowed range of a variable | 04.1 |
| **Candidate** | One choice of values for all variables | 04.1 |
| **Dominated** | A candidate is dominated if another one is at least as good in all objectives and better in one | 04.1 |
| **Optimizer** | The algorithm that proposes candidates and searches for better ones, e.g. `NelderMead`, `U_NSGA3` | 04.1 |
| **Pareto front** | The candidates that are not dominated: improving one objective worsens another | 04.1 |
| **Population / generation** | A set of candidates evaluated together; one iteration of a population-based optimizer | 04.2 |

## Summary

- Parameters belong to the model; variables are the parameters or settings an optimizer changes.
- Objectives rank candidates; constraints decide which ones are acceptable.
- "Converged" can refer to the optimizer, to cyclic stationarity or to the discretization grid.
