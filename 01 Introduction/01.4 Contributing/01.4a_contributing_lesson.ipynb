CADET is developed in the open. Questions, bug reports, examples and code from users improve it for everyone. This lesson shows where each kind of contribution goes and what makes it easy to act on.

## Where to ask what

| You want to … | Use | Where |
|---|---|---|
| ask how to model something, or understand a result | **forum** post | [forum.cadet-web.de](https://forum.cadet-web.de) |
| get help with an error, or report an error or a wrong result | **forum** post or **Issue** | the forum, or the repository of the tool (below) |
| suggest a new feature | **Issue** | the repository of the tool |
| propose a concrete change: a fix, an example, documentation | **pull request** (PR) | the repository of the tool |

Each CADET tool ([01.2 CADET Overview](../01.2%20CADET%20Overview/)) has its own Issue tracker:

| Problems with … | Issues |
|---|---|
| model equations, solver errors, simulation speed | [CADET-Core](https://github.com/cadet/CADET-Core/issues) |
| Python classes, fractionation, fitting, optimization | [CADET-Process](https://github.com/fau-advanced-separations/CADET-Process/issues) |
| the widgets | [CADET-GUI](https://github.com/cadet/CADET-GUI/issues) |
| these notebooks | [CADET-Workshop](https://github.com/cadet/CADET-Workshop/issues) |

Troubleshooting and error reports are welcome on the forum as well; when unsure whether something is a bug, the forum is a good first step.

## Repositories, commits and branches

The CADET projects are managed with **git** and hosted on **GitHub**.

| Term | Meaning |
|---|---|
| **repository** | a project folder together with its complete history |
| **commit** | a saved set of changes with a message, e.g. "Fix capacity check" |
| **branch** | a line of commits that can be developed independently |
| **clone** | a local copy of a repository |
| **fork** | your own copy of a repository on GitHub, to which you can push changes |

```{figure} ../resources/git_branches.svg
:width: 700px
:align: center
:alt: Branches master, dev and a feature branch. The feature branch starts from dev and returns to dev with a pull request; dev is released into master.
```

CADET-Process keeps released versions on `master` and the next release on `dev`. New work happens on a separate branch, such as `feature/new-binding-model` or `fix/capacity-check`, and returns to `dev` through a pull request. The workshop repository has two generated branches instead: `teaching` with empty answer cells and `solution` with completed ones.

## Reporting a problem

A forum post or an Issue is easy to act on when someone else can reproduce the problem. The CADET-Process bug report asks for:

| Field | Content |
|---|---|
| **Description** | what happened, and what you expected instead |
| **Steps to reproduce** | a minimal example (next slide) |
| **Version** | CADET-Process version or commit |
| **Environment** | operating system, Python and CADET-Core versions |
| **Logs or tracebacks** | the complete error message |

The versions of the current environment:

````python
import platform

import CADETProcess
from CADETProcess.simulator import Cadet

print("CADET-Process:", CADETProcess.__version__)
print("CADET-Core:", Cadet().version)
print("Python:", platform.python_version(), "on", platform.system())
````

## Minimal reproducible example

A **minimal reproducible example** is the shortest code that still shows the problem and runs on its own in a fresh kernel.

- Keep the imports and the lines that lead to the problem.
- Remove plots, unrelated unit operations and later steps.
- Replace data files with a few numbers, or attach the file.
- Run it once more in a fresh kernel before posting.

For example, a column model with three components stops with `ValueError: Expected size 3`. The traceback ends inside CADET-Process; its first lines that come from our own code point to the cause. Reduced to four lines, the problem remains:

```python
from CADETProcess.processModel import ComponentSystem, Langmuir

component_system = ComponentSystem(["A", "B", "C"])
binding_model = Langmuir(component_system, name="langmuir")
binding_model.capacity = [10, 10]  # ValueError: Expected size 3
```

The minimal example often reveals the cause: here, `capacity` needs one value per component. It also shows that the message does not name the parameter, which could be suggested as an improvement.

## Proposing a change

A pull request asks the maintainers to merge commits from your branch into theirs.

| Step | On GitHub or in a terminal |
|---|---|
| 1. Fork the repository | **Fork** button on GitHub |
| 2. Clone your fork | `git clone https://github.com/<your-name>/CADET-Process` |
| 3. Create a branch from `dev` | `git switch -c fix/capacity-message dev` |
| 4. Change files and commit | `git add …` then `git commit -m "Name the parameter in size errors"` |
| 5. Push the branch to your fork | `git push origin fix/capacity-message` |
| 6. Open the pull request against `dev` | GitHub offers **Compare & pull request**; mention the Issue, e.g. "Fixes #123" |
| 7. Review and tests | automatic tests run; maintainers comment, then merge |

```{note}
Each project describes its own rules in a `CONTRIBUTING.md` file, e.g. [CADET-Process: contributing](https://github.com/fau-advanced-separations/CADET-Process/blob/dev/CONTRIBUTING.md): branch names, code style, tests and documentation. One pull request should address one Issue or feature.
```

## Ways to contribute

- Answer a question on the forum.
- Report a bug with a minimal example.
- Improve documentation or the workshop notebooks: a typo fix is a good first pull request.
- Share a model or a workflow as an example.
- Collaborate with us: we are open to joint projects. Get in touch on the forum.

## Summary

- Questions, troubleshooting and error reports go to the forum; bugs and feature requests can also go to Issues, concrete changes to pull requests.
- A good error report contains a minimal reproducible example, the versions and the complete traceback.
- Changes are developed on their own branch and merged into `dev` through a pull request.
