A colleague sends a column model that stops with an error and asks whether to report it. Prepare the report as it would be posted.

## Exercise 1: find the cause in a traceback

Run the colleague's model. It stops with an error; this is intended.

````python
# The colleague's model: three components with Langmuir binding
from CADETProcess.processModel import ComponentSystem, Langmuir, LumpedRateModelWithoutPores

component_system = ComponentSystem(["A", "B", "C"])

column = LumpedRateModelWithoutPores(component_system, name="column")
column.length = 0.1
column.diameter = 0.02
column.total_porosity = 0.35
column.axial_dispersion = 1e-7

binding_model = Langmuir(component_system, name="langmuir")
binding_model.is_kinetic = False
binding_model.adsorption_rate = [1, 2, 3]
binding_model.desorption_rate = [1, 1, 1]
binding_model.capacity = [10, 10]
column.binding_model = binding_model
````

Read the traceback from the top. Which line of the cell above causes the error, and what does the last line of the traceback say?

```{dropdown} Answer
The last line says `ValueError: Expected size 3`. The first part of the traceback points to the cell above, at `binding_model.capacity = [10, 10]`; the remaining lines are inside CADET-Process, where the size is checked. The model has three components, so `capacity` needs three values.
```

Work on a branch, then merge it back, as a pull request would:

1. Create and switch to a branch `feature/porosity`: `repo.git("switch", "-c", "feature/porosity")`.
2. Add the line `"Total porosity: 0.35"` with `repo.edit(...)` and commit it: `repo.git("commit", "-am", "Add porosity")`.
3. Switch back to `dev` and merge the branch: `repo.git("switch", "dev")`, then `repo.git("merge", "feature/porosity")`.
4. Show the history.

````python
# Branch, commit and merge
repo.git("switch", "-c", "feature/porosity")
repo.edit("Total porosity: 0.35")
repo.git("commit", "-am", "Add porosity")
repo.git("switch", "dev")
repo.git("merge", "feature/porosity")
repo.history()
````

```{dropdown} Answer
After the merge, `dev` and `feature/porosity` point to the same commit "Add porosity" on top of "Add column notes". Because `dev` had no new commits in the meantime, git simply moves `dev` forward ("Fast-forward"). On GitHub, the same step happens when a maintainer merges a pull request.
```
