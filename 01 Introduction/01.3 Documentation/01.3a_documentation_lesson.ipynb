## Two kinds of documentation

| Documentation | Describes | Look here for |
|---|---|---|
| [CADET-Core](https://cadet.github.io/master/index.html) | the **model equations** | what a model assumes, its symbols and their meaning |
| [CADET-Process](https://cadet-process.readthedocs.io/) | the **Python classes**: user guide, examples and API reference | class names, attribute names, units and methods |

A model appears in both: its equations in CADET-Core, its class in CADET-Process. For example, the [lumped rate model without pores](https://cadet.github.io/master/modelling/unit_operations/lumped_rate_model_without_pores.html) is the class [`LumpedRateModelWithoutPores`](https://cadet-process.readthedocs.io/en/latest/reference/generated/CADETProcess.processModel.LumpedRateModelWithoutPores.html).

## From equation to attribute

The CADET-Core page writes the model with symbols; CADET-Process uses descriptive attribute names in SI units:

| Symbol in CADET-Core | Meaning | Attribute in CADET-Process |
|---|---|---|
| $\varepsilon_t$ | total porosity | `total_porosity` |
| $D_{\text{ax}}$ | axial dispersion coefficient | `axial_dispersion` (m²/s) |
| $L$ | column length | `length` (m) |
| $u$ | velocity, from flow rate and cross-section | inlet `flow_rate` (m³/s) and column `diameter` (m) |
| $c^\ell$, $c^s$ at $t = 0$ | liquid and bound concentrations at the start | initial conditions `c` and `q` (mol/m³) |

Every model object lists its attributes and current values:

```python
from CADETProcess.processModel import ComponentSystem, LumpedRateModelWithoutPores

component_system = ComponentSystem(["Tracer"])
column = LumpedRateModelWithoutPores(component_system, name="column")
print(column.parameters)
print("Required:", column.required_parameters)
```

`None` marks values that still have to be set; `required_parameters` lists them. Lists such as `c` have one entry per component, in the order of the `ComponentSystem`.

## Summary

- CADET-Core documentation describes the model equations; CADET-Process documentation describes the Python classes and their attributes.
- Equation symbols map to descriptive attribute names in SI units; `parameters` and `required_parameters` show them for any model object.
- The catalogs list all available unit operation, binding and reaction models.
