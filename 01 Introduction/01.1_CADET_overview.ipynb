# Naming Conventions 

## Unit Operation Model

```{figure} ./resources/unit_operation.png
:width: 300px
:align: left

```

In bioprocess engineering, a unit operation is a basic step in a process. Unit operations involve a physical change or chemical transformation such as separation, crystallization, filtration, and other reactions. <br>
In CADET unit operations can be for example columns or reactors, but there also inlets and outlets are considered unit operations in a broader sense.

## Flow Sheet Model

```{figure} ./resources/batch_elution.svg
:width: 500px
:align: left

```

The flow sheet model contains information on how different unit operations are connected, including periphery such as pumps, vents and tubing.

## Process Model

```{figure} ./resources/batch_elution_process_model.svg
:width: 800px
:align: left

```

The process model contains information about the duration, cycles and timed events that influence the process as well as the flow sheet model.

## Simulation

