## Lessons and exercises

`02 CADET Fundamentals` is a **chapter**; `02.3 Inlet Profiles and Events` is a **topic** within it. Each topic has a lesson and exercises:

| | Lesson | Exercises |
|---|---|---|
| **Notebook** | `02.3a_inlet_profiles_and_events_lesson.ipynb` | `02.3b_inlet_profiles_and_events_exercises.ipynb` |
| **You** | follow the presentation | write the code yourself, alone or in pairs |
| **Code** | shown and typed live by the presenter | empty answer cells for you to fill |

You do not need to type along during lessons: everything typed live is filled in on the `solution` branch. Your own coding starts in the exercises. Those marked **optional** extend the main workflow and can be skipped.

Stuck or behind? Each topic starts from a ready-made state, so you can always continue with the next lesson.

## Teaching and solutions

The workshop lives in one repository, [CADET-Workshop](https://github.com/cadet/CADET-Workshop), with two branches generated from the same source, so cells line up one to one:

| Branch | Contains | On the JupyterHub |
|---|---|---|
| `teaching` | lessons and exercises, solution cells empty | `CADET-Workshop-2026-teaching`: work here |
| `solution` | the same notebooks, all solution cells filled in | `CADET-Workshop-2026-solution`: check your work |

The solutions are also on [GitHub](https://github.com/cadet/CADET-Workshop/tree/solution), or in a local copy with `git checkout solution`. Exercises also have **Answer** dropdowns below them, to compare your reasoning.

## Working in JupyterLab

| Name | What it is |
|---|---|
| **Jupyter notebook** | a document (`.ipynb`) of text and code cells; code runs in a Python **kernel** and its output appears below the cell |
| **JupyterLab** | the program in the browser in which notebooks are opened, edited and run, with a file browser and terminal |
| **JupyterHub** | a server that starts a JupyterLab for each user, so nothing has to be installed; the workshop runs on [jupyter.cadet-web.de](https://jupyter.cadet-web.de) |

- **Shift+Enter** runs a cell and moves on; **Ctrl+Enter** runs it and stays.
- Run cells **from top to bottom**: later cells use what earlier cells created.
- `[3]` next to a cell: it was the third cell run; `[*]`: it is still running.
- Start over: *Kernel → Restart Kernel and Run All Cells*.
- Collapsed cells: click the bar on their left.

## Finding your way in the files

- The **file browser** (folder icon in the left side bar) shows the workshop folders: chapter → topic → notebooks.
- A **double-click** on an `.ipynb` file opens it in a new tab.
- Dragging a tab to the right edge shows two notebooks side by side, e.g. a lesson and its exercises.
- The links at the end of each notebook open its exercises and the next lesson directly.

:::{dropdown} Starting over with a fresh copy of the workshop files
On the JupyterHub, the Launcher (*File → New Launcher*) has an **Update workshop files** card at the top. Close the notebooks from the workshop folders first, since open notebooks keep pointing to the old files. The card opens a terminal that asks for each workshop folder what to do:

- **Archive** (default): renames the folder to `<folder>_archive_<date>_<time>`, so your own changes stay there.
- **Overwrite:** deletes the folder, after you type `yes`.
- **Skip:** leaves the folder as it is.

It then downloads a fresh copy of every folder that was archived or overwritten.
:::
