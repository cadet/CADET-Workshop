
CADET is a family of open-source tools for modeling and simulating chromatography and related processes. All repositories are on [github.com/cadet](https://github.com/cadet/).

## Purpose of modeling

A mechanistic model describes a process by the physical and chemical effects behind it. Once its parameters are known, it can simulate the process under conditions that were not measured.

| Question | Examples |
|---|---|
| What happens inside a unit operation? | convection, dispersion, film and pore diffusion, binding |
| How do the units work together? | flow sheet connections, system periphery such as mixers and tubing |
| How does the process change over time? | elution gradients, column switching, cyclic operation |
| Which parameters describe the system? | characterization: estimating transport and binding parameters from experiments |
| How well does the process perform? | purity, recovery and productivity |
| Which operating conditions are best? | process optimization, e.g. of the gradient, the load or the cycle time |

## The CADET tools

| Tool | What it is | Used for | Links |
|---|---|---|---|
| **CADET-Core** | the solver, written in C++ | solving the model equations, fast and accurate | [repository](https://github.com/cadet/CADET-Core/), [documentation](https://cadet.github.io/master/index.html) |
| **CADET-Python** | a thin Python interface to CADET-Core | writing and running CADET-Core input files | [repository](https://github.com/cadet/CADET-Python) |
| **CADET-Process** | a Python toolbox for modeling and optimization | building processes, simulating, fractionating, fitting and optimizing; **most of this workshop** | [repository](https://github.com/fau-advanced-separations/CADET-Process), [documentation](https://cadet-process.readthedocs.io/) |
| **CADET-GUI** | interactive widgets on top of CADET-Process | exploring models in the notebook, without code | [repository](https://github.com/cadet/CADET-GUI) |
| **CADET-RDM** | research data management | tracking versions and metadata of simulation studies | [repository](https://github.com/cadet/CADET-RDM), [documentation](https://cadet-rdm.readthedocs.io/) |

## How the tools connect

```{figure} resources/cadet_overview.png
:width: 800px
:align: center
```

## From equipment to a simulation

**Components → unit operations → flow sheet → process → simulation**

| Object | What it describes |
|---|---|
| `ComponentSystem` | Species in the experiment |
| Unit operations | Equipment and its model parameters |
| `FlowSheet` | Connections between units |
| `Process` | Flow sheet, cycle time and timed events |
| `Cadet` | Simulator that calculates the concentration profiles |

## Summary

- CADET-Process builds process models and connects simulations, fitting and performance evaluation.
- CADET-Core solves the model equations; CADET-GUI provides interactive access to model settings.
- A flow sheet defines connections; a process adds cycle time and events.
