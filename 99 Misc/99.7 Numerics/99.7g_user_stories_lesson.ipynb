## Fail compilation

Here, we discuss different scenarios of crashing simulations and how to proceed in such cases.

Every case below is taken from a real post on the
[CADET forum](https://forum.cadet-web.de/), reproduced against the current
CADET-Core, and worked through from the error message to the cause. They are
independent of each other, so pick whichever one you want to see.

The table deliberately tells you only what the *user* saw. Finding out what it
really was is the exercise.
