## Selecting the model

**CADET-Core covers a vast model family** with various features and levels of resolution.
**[CADET-Equations](https://equations.cadet-web.de/)** is a web app to explore available models and their governing equations, assumptions and more. The **[CADET-Core Documentation](https://cadet.github.io/master/index.html)** describes how models must be configured for the simulator.

- Models are described deterministically by a system of **Partial differential (algebraic) equations (PD(A)E)**.
- The implementation unifies **various features** in a **modular** manner:
    - **Transport:** Convection, dispersion, film and pore diffusion
    - **Dimensionality:** 1D, 2D, 3D -> different (apparent) dispersion, bed resolution
    - **Particle Model**: resolution of film-, pore-diffusion, porosity, particle-size distribution, ...
    - **Binding:** Variety of sorption mechanisms: ion-exchange, colloidal, hydrophobicity, ...
    - **Reactions:** Mass-action law, Michaelis Menten
    - **Column Geometry:** Axial-flow, radial-flow, conical columns including variable velocity profile, dispersion and film diffusion dependence

## Solving the model

- Analytical solutions are generally not available, we need to rely on **numerical simulation**
- **Method of lines** is used to solve the governing PDAE:
    - Separate, **sequential discretization of spatial and time coordinate**
    - Spatial discretization yields an ODAE 
    - Implicit temporal discretization yields a **nonlinear algebraic system** to be solved e.g. by Newtons method

## Using CADET - Choosing Numerical Parameters

- CADET users must **select suitable parameters** to obtain **accurate** and **efficient simulations**.
- Depending on the model and numerical method, **numerous parameters** must be configured:
    - Spatial Discretization: `SPATIAL_METHOD`, `NCOL`, `NPAR`, `WENO_ORDER`, `RECONSTRUCTION`, `INTEGRATION_TYPE`, ...
    - Temporal Discretization: `ABSTOL`, `RELTOL`, `ALGTOL`, ...
    - Initial value configuration: `CONSISTENT_INIT_MODE`, ...
    - Nonlinear solver configuration: `GS_TYPE`, `MAX_KRYLOV`, `MAX_RESTARTS`, `SCHUR_SAFETY`, ...


Some have **defaults** that usually work, but **some are very problem-specific**.
This raises several practival questions:
